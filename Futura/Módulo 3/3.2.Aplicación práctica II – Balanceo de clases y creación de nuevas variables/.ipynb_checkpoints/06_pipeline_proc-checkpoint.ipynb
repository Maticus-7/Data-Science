## 6.0 El catálogo de Feature Engineering

Antes del código, el **menú**. El algoritmo es una commodity; las features salen del
**conocimiento del negocio**. Estos son los tipos de variables derivadas que un científico de
datos crea, con ejemplos de **este** problema y de **otras casuísticas** para que veas el patrón.

| Tipo | Qué es | Aquí (crédito) | Otras casuísticas |
|---|---|---|---|
| **Ratios** | a / b | `%Cuota` = cuota/ingreso (endeudamiento) | Churn: minutos_uso/minutos_plan · Fraude: monto/límite |
| **Agregaciones** | suma/media/máx de varias | `TotalIncome` = solicitante+co | Retail: gasto_medio_6m · Banca: saldo_promedio_mes |
| **Flags binarios** | ¿ocurre la condición? | `Tiene_Coaplicante` (0 ≠ ausencia) | Fraude: opera_de_madrugada · Churn: uso_cayó_50% |
| **Interacciones** | a × b, a condicional a b | ingreso × plazo | Marketing: edad × canal · Riesgo: mora × antigüedad |
| **Diferencias / deltas** | a − media(grupo) | ingreso − mediana(zona) | Series: consumo_hoy − consumo_ayer |
| **Transformaciones no lineales** | log, √, potencia | `log(ingreso)` (des-sesga) | Cualquier variable con cola larga |
| **Binning / discretización** | continua → tramos | tramos de ingreso | Scorecards, WoE (NB05) |
| **Temporales** | recencia, frecuencia, tendencia | (no hay fecha aquí) | RFM: recencia/frecuencia/monto |
| **Basadas en target** | media del target por categoría | *(cuidado: leakage — solo con CV)* | Target/mean encoding |
| **Conteos / cardinalidad** | nº de eventos, nº de productos | nº de dependientes | Nº de reclamos, nº de logins |

> 🎓 **Regla de oro del FE:** cada variable derivada debe tener una **hipótesis de negocio**
> detrás. `%Cuota` no es "ApplicantIncome dividido LoanAmount": es *"qué parte del sueldo se va
> en pagar la deuda"*. Si no puedes explicar por qué una feature debería predecir, probablemente
> es ruido con suerte.

A continuación construimos una selección de estas para nuestro caso (sección 6.5).

# 6. Pipeline de preprocesamiento

Llegamos al corazón de la sesión. Vamos a transformar el dataset crudo en una matriz numérica
lista para modelar, aplicando **en el orden correcto**:

```
  1. Labeling            (categórica ordinal → numérica)
  2. Indicadores de nulo (marcar antes de destruir la evidencia)
  3. Imputación          (fit en TRAIN)
  4. Feature engineering (crear variables de negocio)
  5. Tratamiento de atípicos (percentiles de TRAIN)
  6. Transformaciones no lineales (log)
  7. Encoding            (categorías de TRAIN)
  8. Escalamiento        (fit en TRAIN)
```

## ⚠️ El principio que rige todo este bloque

| Conjunto | Operación | Método |
|---|---|---|
| **TRAIN** | Aprende los parámetros **y** los aplica | `.fit_transform()` |
| **TEST** | **Solo** aplica los parámetros ya aprendidos | `.transform()` |

> 🔴 **El error #1 en notebooks de preprocesamiento** es escribir `scaler.fit_transform(X_test)`.
> Eso hace que el test se escale con **sus propios** mínimos y máximos, no con los del train.
> Resultado: (a) el modelo recibe en producción datos en una escala distinta a la que entrenó,
> y (b) las métricas de test quedan **optimistamente sesgadas**, porque el test usó información
> que no debía conocer. En producción **no existe** un "conjunto de test" del cual aprender:
> los clientes llegan **de a uno**. Si tu preprocesamiento necesita ver el lote completo
> para funcionar, **no es desplegable**.

Para hacer esto explícito y auditable, guardaremos todo lo aprendido en un diccionario
`params_aprendidos`. Al final, ese diccionario **es** el modelo de preprocesamiento: es lo
único que hay que serializar y llevar a producción.

````python
# Aquí vive TODO lo que se aprende de los datos. Nada fuera de este dict puede tocar el test.
params_aprendidos = {}

# Trabajamos sobre copias para que el notebook sea IDEMPOTENTE:
# re-ejecutar una celda no debe cambiar el resultado.
Xtr = X_train.copy()
Xte = X_test.copy()

print('Train:', Xtr.shape, '| Test:', Xte.shape)
````

## 6.1 Labeling — variables categóricas ordinales

Para convertir una variable categórica **ordinal** a numérica se utiliza **labeling**.
Se usa cuando existe una **noción de distancia/orden** entre las categorías.

> ⚠️ Se debe tener cuidado en que el orden de los números utilizado sea **el mismo orden**
> que representan las categorías originales. Si asignas etiquetas arbitrarias a una variable
> **nominal**, le estás inventando un orden al modelo (le dices que `Rural(0) < Urban(2)`,
> y que la distancia Rural→Urban es el doble que Rural→Semiurban). Eso es un error conceptual
> grave: para nominales se usa **one-hot encoding** (sección 6.7).

````python
Xtr['Dependents'].value_counts(dropna=False)
````

`Dependents` **sí** es ordinal: `0 < 1 < 2 < 3+`. El labeling es apropiado.

Nota sobre `'3+'`: es un **techo** (censura). Al mapearlo a `3` asumimos que "3 o más" ≈ 3.
Es una simplificación aceptable porque el grupo es pequeño, pero hay que **saber que se está
haciendo**.

````python
mapping_dependents = {'0': 0, '1': 1, '2': 2, '3+': 3}
params_aprendidos['mapping_dependents'] = mapping_dependents

# Se aplica el MISMO mapping a train y test.
Xtr['Dependents'] = Xtr['Dependents'].map(mapping_dependents)
Xte['Dependents'] = Xte['Dependents'].map(mapping_dependents)

print('TRAIN:'); print(Xtr['Dependents'].value_counts(dropna=False).sort_index())
print()
print('TEST :'); print(Xte['Dependents'].value_counts(dropna=False).sort_index())
````

> 🔎 **Detalle fino:** `.map()` convierte a `NaN` cualquier valor que no esté en el diccionario.
> Los nulos originales siguen siendo nulos (bien), pero si en test apareciera una categoría
> nueva (p. ej. `'4+'`), se volvería `NaN` **silenciosamente**. En producción esto hay que
> **loguearlo**, no ignorarlo.

````python
# Validación defensiva: ¿el mapeo introdujo nulos que no existían antes?
nuevos_tr = Xtr['Dependents'].isnull().sum() - X_train['Dependents'].isnull().sum()
nuevos_te = Xte['Dependents'].isnull().sum() - X_test['Dependents'].isnull().sum()
print(f'Nulos NUEVOS introducidos por el mapeo → train: {nuevos_tr} | test: {nuevos_te}')
assert nuevos_tr == 0 and nuevos_te == 0, 'El mapeo de Dependents introdujo nulos inesperados'
print('OK: el mapeo no introdujo categorías desconocidas.')
````

## 6.2 `Credit_History`: cuando el nulo ES información

`Credit_History` viene como `0.0` / `1.0`, así que pandas la lee como **numérica**. El notebook
ingenuo la imputaría con la **mediana** junto al resto de numéricas. Eso es un error doble:

1. Es una variable **binaria/categórica**, no continua. Su "mediana" (1.0) no tiene
   interpretación como valor central: es simplemente la categoría más frecuente.
2. Vimos en el diagnóstico del NB02 (EDA) que **la tasa de rechazo de los nulos es muy
   distinta** de la de los no-nulos. El nulo **no es aleatorio (MNAR)**: probablemente
   corresponde a clientes **sin historial en el sistema financiero** (nuevos, no bancarizados).
   Imputarlos como `1` (buen historial) **destruye esa señal** y, peor, los disfraza de
   clientes buenos.

**Decisión:** la tratamos como categórica de **tres niveles**, con el desconocido
como categoría propia.

````python
tabla = pd.DataFrame({
    'n': X_train['Credit_History'].isnull().map({True: 'NULO', False: 'No nulo'}).value_counts(),
})
tabla['tasa_rechazo'] = y_train.groupby(
    X_train['Credit_History'].isnull().map({True: 'NULO', False: 'No nulo'})).mean()
tabla['tasa_rechazo'] = tabla['tasa_rechazo'].map('{:.1%}'.format)
print('Comportamiento del target según Credit_History esté o no informado:')
print(tabla)
print(f'\n(tasa global: {y_train.mean():.1%})')
print('\n>> Si las tasas difieren, el nulo es informativo: NO debe imputarse a ciegas.')
````

````python
def tratar_credit_history(df):
    """Convierte Credit_History en categórica de 3 niveles, preservando el nulo como categoría."""
    out = df.copy()
    out['Credit_History'] = (out['Credit_History']
                             .map({0.0: 'Sin_historial_OK', 1.0: 'Historial_OK'})
                             .fillna('Desconocido'))
    return out

Xtr = tratar_credit_history(Xtr)
Xte = tratar_credit_history(Xte)

print('TRAIN:'); print(Xtr['Credit_History'].value_counts())
print()
print('TEST :'); print(Xte['Credit_History'].value_counts())
````

## 6.3 Indicadores de nulo (*missing indicators*)

Antes de imputar, **marcamos**. La imputación rellena el hueco y borra la evidencia de que
alguna vez hubo un hueco. Si esa ausencia era informativa (MNAR), acabamos de perder señal.

Un indicador binario `<variable>_missing` conserva esa información **sin costo**: el modelo
decide si la usa o no.

> Del marco conceptual: *"¿Cómo lidiar con valores extremos? — Eliminarlos: se puede perder
> información. **Marcarlos: opción más segura.**"* — el mismo principio aplica a los nulos.

````python
# Aprendemos en TRAIN qué variables tienen nulos. Test usa exactamente esta misma lista,
# aunque en test alguna no tenga nulos (si no, las columnas no coincidirían).
cols_con_nulos = [c for c in Xtr.columns if Xtr[c].isnull().any()]
params_aprendidos['cols_con_nulos'] = cols_con_nulos
print('Variables con nulos en TRAIN (se les creará indicador):', cols_con_nulos)
````

````python
def agregar_indicadores_nulo(df, columnas):
    out = df.copy()
    for c in columnas:
        out[f'{c}_missing'] = out[c].isnull().astype(int)
    return out

Xtr = agregar_indicadores_nulo(Xtr, cols_con_nulos)
Xte = agregar_indicadores_nulo(Xte, cols_con_nulos)

print('Shape train:', Xtr.shape, '| test:', Xte.shape)
Xtr[[f'{c}_missing' for c in cols_con_nulos]].sum().to_frame('n_nulos_marcados')
````

## 6.4 Imputación de datos faltantes

> *"El método de imputación depende de la naturaleza de la variable."*

Las dos preguntas guía del marco conceptual:

- *Si hay valores nulos en la **edad** de clientes, ¿qué valor debería imputar?*
  → Una edad **existe** siempre; el dato simplemente no se capturó. Tiene sentido estimarla:
  mediana, o mejor, la mediana **del segmento** al que pertenece.
- *Si hay valores nulos en la **cantidad de transacciones** de un cliente, ¿qué valor debería
  imputar?* → Aquí el nulo casi siempre significa **cero**: el cliente no transó, por eso no
  aparece en la tabla de transacciones. Imputar la mediana le inventaría actividad que no tuvo.

> # **La moraleja: no existe "la" estrategia de imputación. Existe entender de dónde viene el dato.**

| Estrategia | Cuándo usarla | Riesgo |
|---|---|---|
| **Media** | Numérica simétrica, sin atípicos | Se distorsiona con colas largas |
| **Mediana** | Numérica asimétrica / con atípicos | Reduce la varianza real |
| **Moda** | Categórica | Sobre-representa la categoría dominante |
| **Constante** (0, "Desconocido") | El nulo tiene significado de negocio | Requiere conocer el dominio |
| **KNN / iterativa** | Nulos MAR, variables correlacionadas | Costoso; puede filtrar información |

Separamos numéricas de categóricas, porque cada grupo recibe un tratamiento distinto.

````python
from sklearn.impute import SimpleImputer

# Las listas de columnas se DEFINEN EN TRAIN y se reutilizan en test. Nunca se recalculan
# sobre test: si lo hiciéramos y test tuviera otros dtypes, las columnas se desalinearían.
cols_num = Xtr.select_dtypes('number').columns.tolist()
cols_cat = Xtr.select_dtypes('object').columns.tolist()

params_aprendidos['cols_num'] = cols_num
params_aprendidos['cols_cat'] = cols_cat

print('NUMÉRICAS  :', cols_num)
print()
print('CATEGÓRICAS:', cols_cat)
````

### ¿Media o mediana? Veamos por qué importa

````python
comparacion = pd.DataFrame({
    'media': Xtr[cols_num].mean(),
    'mediana': Xtr[cols_num].median(),
})
comparacion['ratio_media/mediana'] = (comparacion['media'] / comparacion['mediana']).round(2)
comparacion['asimetría (skew)'] = Xtr[cols_num].skew().round(2)
print(comparacion)
print()
print('>> ApplicantIncome tiene una media MUY superior a su mediana y una asimetría alta:')
print('   unos pocos ingresos enormes arrastran la media. Imputar con la media le regalaría')
print('   a un cliente sin dato un ingreso que NO es representativo. Usamos la MEDIANA.')
````

### Imputación de numéricas — `fit` en TRAIN

````python
imputer_num = SimpleImputer(strategy='median')

# fit SOLO en train  ← aquí se aprenden las medianas
imputer_num.fit(Xtr[cols_num])

params_aprendidos['imputer_num'] = imputer_num

print('Medianas aprendidas EN TRAIN (esto es lo que va a producción):')
print(pd.Series(imputer_num.statistics_, index=cols_num).to_string())
````

````python
# Aplicamos a AMBOS conjuntos, pero con los parámetros aprendidos SOLO en train.
# ⚠️ Preservamos el ÍNDICE original: si no lo hacemos, el posterior pd.concat con las
#    categóricas mezclaría filas de clientes distintos SIN AVISAR. Bug silencioso y letal.
Xtr_num = pd.DataFrame(imputer_num.transform(Xtr[cols_num]),
                       columns=cols_num, index=Xtr.index)
Xte_num = pd.DataFrame(imputer_num.transform(Xte[cols_num]),   # <-- transform, NO fit_transform
                       columns=cols_num, index=Xte.index)

print('Nulos restantes → train:', Xtr_num.isnull().sum().sum(), '| test:', Xte_num.isnull().sum().sum())
print('Índice preservado en train:', Xtr_num.index.equals(Xtr.index))
print('Índice preservado en test :', Xte_num.index.equals(Xte.index))
Xtr_num.head()
````

### Imputación de categóricas — `fit` en TRAIN

````python
imputer_cat = SimpleImputer(strategy='most_frequent')
imputer_cat.fit(Xtr[cols_cat])

params_aprendidos['imputer_cat'] = imputer_cat

print('Modas aprendidas EN TRAIN:')
print(pd.Series(imputer_cat.statistics_, index=cols_cat).to_string())
````

````python
Xtr_cat = pd.DataFrame(imputer_cat.transform(Xtr[cols_cat]),
                       columns=cols_cat, index=Xtr.index)
Xte_cat = pd.DataFrame(imputer_cat.transform(Xte[cols_cat]),
                       columns=cols_cat, index=Xte.index)

print('Nulos restantes → train:', Xtr_cat.isnull().sum().sum(), '| test:', Xte_cat.isnull().sum().sum())
Xtr_cat.head()
````

### 🔴 Demostración del leakage: `fit_transform` vs `transform` en test

Veamos con números **cuánto cambia** el resultado si cometemos el error clásico.

````python
imputer_malo = SimpleImputer(strategy='median')
imputer_malo.fit(Xte[cols_num])   # ❌ aprende del TEST

comparacion_leak = pd.DataFrame({
    'mediana_TRAIN (correcto)': imputer_num.statistics_,
    'mediana_TEST (leakage)': imputer_malo.statistics_,
}, index=cols_num)
comparacion_leak['diferencia'] = (comparacion_leak['mediana_TEST (leakage)']
                                  - comparacion_leak['mediana_TRAIN (correcto)'])
comparacion_leak['dif_%'] = (comparacion_leak['diferencia']
                             / comparacion_leak['mediana_TRAIN (correcto)'] * 100).round(2)
print('Si hubiéramos hecho fit en TEST, imputaríamos valores DISTINTOS:')
print(comparacion_leak)
print()
print('>> Cada fila con diferencia != 0 es una fuga de información del test hacia el modelo.')
print('   Y sobre todo: en PRODUCCIÓN estos valores NO EXISTEN. El cliente llega de a uno;')
print('   no hay un "lote de test" del cual calcular una mediana. El código no sería desplegable.')
````

## 6.5 Creación de nuevas variables (Feature Engineering)

<img src='https://i.pinimg.com/originals/89/0f/92/890f92f1dba99065531aa2bc1b6b9c45.jpg' width='420'>

> *"Usualmente las variables que tenemos disponibles en nuestra base de datos **no son
> suficientes** para el desarrollo de nuestro modelo predictivo. ¿Podemos crear nuevas
> variables a partir de las variables iniciales?"*

**Aquí es donde el científico de datos genera el valor diferencial.** El algoritmo lo puede
elegir cualquiera; las features salen del **conocimiento del negocio**.

Ejemplos del marco conceptual:
- Promedio de saldo por mes en los últimos *x* meses.
- ¿Nuestro cliente recibe su sueldo en soles o dólares?
- ¿Suele comprar con su tarjeta de crédito en soles o dólares?
- ¿Paga mensualmente algún crédito en dólares?

Tipos de transformación:

| Tipo | Operación | Ejemplo aquí |
|---|---|---|
| **Ratios** | a / b | `%Cuota` = cuota / ingreso → **nivel de endeudamiento** |
| **Agregaciones** | a + b | `TotalIncome` = ingreso solicitante + co-solicitante |
| **Desviaciones** | a − media(grupo) | ingreso vs. la media de su zona |
| **Transformaciones lineales** | escalado, centrado | estandarización |
| **Transformaciones NO lineales** | log, exp, raíz | `log(ingreso)` |

Las que vamos a construir tienen **sentido de negocio explícito** — un analista de riesgo las
reconocería de inmediato:

````python
def construir_features(df_num, df_cat, mediana_cuota=None):
    """
    Crea variables derivadas con significado de negocio.

    Esta función es DETERMINÍSTICA salvo por UN parámetro: `mediana_cuota`. Y precisamente
    por eso ese parámetro se recibe DESDE FUERA, en vez de calcularse aquí dentro.

    Si escribiéramos `X['Cuota'].fillna(X['Cuota'].median())`, la función aprendería del
    DataFrame que le toque: al llamarla con el test, el test se imputaría con SU PROPIA
    mediana. Sería exactamente el error que denunciamos en 6.4 y 6.9, escondido dentro
    de una función con un comentario que dice "determinística".

    Regla: si una función necesita un parámetro aprendido, que lo reciba. Así el parámetro
    queda visible en `params_aprendidos` y es auditable.
    """
    X = df_num.copy()
    eps = 1e-6   # evita divisiones por cero

    # 1) Ingreso total del hogar: el banco evalúa la capacidad de pago CONJUNTA.
    X['TotalIncome'] = X['ApplicantIncome'] + X['CoapplicantIncome']

    # 2) Ingreso per cápita: no es lo mismo ganar 5000 solo que 5000 con 3 dependientes.
    X['TotalIncomeByPerson'] = X['TotalIncome'] / (X['Dependents'] + 1)

    # 3) Cuota mensual estimada. LoanAmount viene en MILES → x1000.
    #    Guarda contra Loan_Amount_Term = 0 (un plazo de cero meses no existe, pero el dato
    #    sucio sí: en producción llegará, aunque hoy este dataset no lo tenga).
    X['Cuota'] = X['LoanAmount'] * 1000 / X['Loan_Amount_Term'].replace(0, np.nan)
    if mediana_cuota is None:                        # sólo ocurre en TRAIN
        mediana_cuota = X['Cuota'].median()
    X['Cuota'] = X['Cuota'].fillna(mediana_cuota)    # en TEST llega la mediana de TRAIN

    # 4) ⭐ RATIO CUOTA/INGRESO: el indicador más usado en riesgo crediticio.
    #    Mide qué porcentaje del ingreso se va en pagar la deuda.
    X['%Cuota'] = X['Cuota'] / (X['TotalIncome'] + eps)

    # 5) Ratio monto solicitado / ingreso: cuántos "sueldos" pide prestado.
    X['LoanToIncome'] = X['LoanAmount'] * 1000 / (X['TotalIncome'] + eps)

    # 6) ¿Hay co-solicitante? Un flag binario suele capturar mejor la señal que el monto crudo,
    #    porque muchos registros tienen CoapplicantIncome = 0 (no es un ingreso bajo:
    #    es la AUSENCIA de co-solicitante). El % exacto se imprime más abajo.
    X['Tiene_Coaplicante'] = (X['CoapplicantIncome'] > 0).astype(int)

    # 7) Proporción del ingreso aportada por el co-solicitante.
    X['Ratio_Coaplicante'] = X['CoapplicantIncome'] / (X['TotalIncome'] + eps)

    # 8) Plazo largo (>= 30 años): flag de estructura del crédito.
    X['Plazo_Largo'] = (X['Loan_Amount_Term'] >= 360).astype(int)

    return X, mediana_cuota


# TRAIN aprende la mediana; TEST la recibe. Nunca al revés.
Xtr_fe, mediana_cuota = construir_features(Xtr_num, Xtr_cat)
Xte_fe, _ = construir_features(Xte_num, Xte_cat, mediana_cuota=mediana_cuota)

params_aprendidos['mediana_cuota'] = mediana_cuota
print(f'Mediana de Cuota aprendida en TRAIN: {mediana_cuota:,.2f}  '
      f'(guardada en params_aprendidos)')

print('Variables tras feature engineering:', Xtr_fe.shape[1], '(antes:', Xtr_num.shape[1], ')')
print('\nNuevas variables creadas:')
print([c for c in Xtr_fe.columns if c not in Xtr_num.columns])
Xtr_fe.head()
````

### 🔎 El detalle que casi todos pasan por alto: `CoapplicantIncome = 0`

````python
pct_cero = (Xtr_num['CoapplicantIncome'] == 0).mean()
print(f'% de registros con CoapplicantIncome = 0 : {pct_cero:.1%}')
print()
print('Ese 0 NO significa "gana muy poco". Significa "NO HAY co-solicitante".')
print('Son dos cosas conceptualmente distintas que comparten el mismo valor numérico.')
print('Por eso creamos el flag Tiene_Coaplicante: separa la AUSENCIA del MONTO.')
print()
print('Tasa de rechazo según tenga o no co-solicitante:')
print(y_train.groupby(Xtr_fe['Tiene_Coaplicante'].map({0: 'Sin co-solicitante', 1: 'Con co-solicitante'}))
      .agg(n='count', tasa_rechazo='mean').assign(tasa_rechazo=lambda d: d['tasa_rechazo'].map('{:.1%}'.format)))
````

### Validación: ¿las features nuevas son sanas?

````python
print('¿Hay infinitos o nulos en las variables derivadas?')
chequeo = pd.DataFrame({
    'n_inf': np.isinf(Xtr_fe).sum(),
    'n_nan': Xtr_fe.isnull().sum(),
    'min': Xtr_fe.min().round(3),
    'max': Xtr_fe.max().round(3),
})
print(chequeo)
assert not np.isinf(Xtr_fe).any().any(), 'Hay valores infinitos en train'
assert not np.isinf(Xte_fe).any().any(), 'Hay valores infinitos en test'
print('\nOK: sin infinitos. Las guardas contra división por cero funcionaron.')
````

## 6.6 Tratamiento de valores atípicos (outliers)

<img src='https://miro.medium.com/max/1352/1*xsJKdRtENPJn4WWx604LGQ.png' width='560'>

> *"Los valores atípicos pueden tener efectos negativos en el entrenamiento de modelos,
> principalmente aquellos que utilizan el **cálculo de distancias** para el aprendizaje."*
> (KNN, K-Means, SVM, y también la regresión lineal/logística vía sus coeficientes.)

### ¿Cómo lidiar con valores extremos?

| Opción | Comentario |
|---|---|
| **Eliminarlos** | Se puede perder información — y si el atípico es real, pierdes justo el caso interesante |
| **Marcarlos** | ✅ **Opción más segura** — creas un flag y conservas todo |
| **Reescalar / capear** | Reduce el efecto del extremo sin perder la fila |

### ¿Cómo identificarlos?

**Validación de valores correctos** (lo primero, y lo que se salta todo el mundo):
- Validación de distribuciones **con los equipos de negocio**
- Validación de que las variables tengan los **rangos adecuados**

**Y de forma automática:** observar los percentiles extremos —
mínimos: `0, 2.5, 5, 10` · máximos: `90, 95, 97.5, 99, 100`

````python
Xtr_fe.describe(percentiles=[0.01, 0.025, 0.05, 0.10, 0.5, 0.90, 0.95, 0.975, 0.99]).T
````

> 📌 **Lectura:** en `ApplicantIncome`, el máximo está órdenes de magnitud por encima del
> percentil 99. Un puñado de solicitantes millonarios va a dominar cualquier cálculo basado
> en distancias o en sumas de cuadrados.
>
> **Pregunta de negocio antes de tocar nada:** ¿un ingreso de 81.000 es un **error de captura**
> (el cliente puso el ingreso anual en un campo mensual) o es un **cliente de banca premium
> real**? La respuesta cambia la decisión: si es error → corregir/eliminar; si es real →
> capear o segmentar. **Esto se pregunta al negocio, no se adivina desde el notebook.**

### Capping (winsorización) por percentiles — aprendidos en TRAIN

````python
# Variables continuas sesgadas que vamos a capear. Los flags binarios y las variables
# discretas NO se capean (no tendría sentido: un flag no tiene "cola").
cols_capear = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount',
               'TotalIncome', 'TotalIncomeByPerson', 'Cuota', '%Cuota', 'LoanToIncome']

# ⚠️ Los percentiles se calculan SOLO sobre TRAIN.
limites = {}
for c in cols_capear:
    limites[c] = (np.percentile(Xtr_fe[c], 1), np.percentile(Xtr_fe[c], 99))

params_aprendidos['limites_capping'] = limites

print('Límites p1 / p99 aprendidos EN TRAIN:')
print(pd.DataFrame(limites, index=['p1', 'p99']).T.round(3))
````

````python
def aplicar_capping(df, limites):
    """np.clip limita los bordes del intervalo a unos valores dados."""
    out = df.copy()
    for c, (lo, hi) in limites.items():
        out[c] = np.clip(out[c], lo, hi)
    return out

# Los MISMOS límites de train se aplican a test.
Xtr_cap = aplicar_capping(Xtr_fe, limites)
Xte_cap = aplicar_capping(Xte_fe, limites)   # <-- límites de TRAIN, no de test

print('Capping aplicado.')
print()
print('% de registros modificados en TRAIN por variable:')
for c in cols_capear:
    pct = (Xtr_fe[c] != Xtr_cap[c]).mean()
    print(f'  {c:24s} {pct:.2%}')
````

> 🔎 **Por qué la función devuelve una copia y no modifica in-place:** el notebook original
> hacía `X['ApplicantIncome'] = np.clip(X['ApplicantIncome'], p1, p99)` directamente sobre el
> DataFrame. Eso funciona **la primera vez**. Pero si en clase re-ejecutas la celda después de
> haber recalculado `p1`/`p99`, capeas **sobre lo ya capeado**: el resultado depende de cuántas
> veces ejecutaste la celda. En una demo en vivo, con alumnos mirando, ese es exactamente
> el momento en que los números dejan de cuadrar y nadie entiende por qué.
>
> **Un notebook debe ser idempotente.** Copias, no mutaciones.

````python
fig, axes = plt.subplots(2, 2, figsize=(13, 7))
for i, col in enumerate(['ApplicantIncome', 'LoanAmount']):
    sns.histplot(Xtr_fe[col], ax=axes[0, i], kde=True, color='#e76f51')
    axes[0, i].set_title(f'{col} — ANTES del capping')
    axes[0, i].set_xlabel('')
    sns.histplot(Xtr_cap[col], ax=axes[1, i], kde=True, color='#2a9d8f')
    axes[1, i].set_title(f'{col} — DESPUÉS del capping (p1–p99)')
    axes[1, i].set_xlabel('')
plt.suptitle('Efecto del capping sobre la distribución', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

### Transformaciones no lineales: el logaritmo

Aplicamos la transformada logaritmo para **normalizar** variables con colas largas y reducir
la escala de valores muy grandes. Una distribución más simétrica ayuda a los modelos lineales,
que asumen relaciones aditivas.

> ⚠️ **Usamos `np.log1p(x) = log(1+x)`, no `np.log(x)`.**
> `CoapplicantIncome` tiene una parte importante de ceros (el % exacto se imprime abajo) y
> `log(0) = -inf`. Un solo `-inf` rompe el escalado, el PCA y el entrenamiento entero — y el
> error aparece 20 celdas después, en un lugar que no tiene nada que ver. `log1p` está
> definida en 0 y devuelve 0.

````python
print('Demostración de por qué np.log es peligroso aquí:')
print('  np.log(0)   =', np.log(0))
print('  np.log1p(0) =', np.log1p(0))
print()
print('Ceros en CoapplicantIncome (train):', int((Xtr_cap['CoapplicantIncome'] == 0).sum()))
print('>> np.log() habría generado ese mismo número de -inf.')
````

````python
def agregar_logs(df):
    out = df.copy()
    for c in ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'TotalIncome', 'Cuota']:
        out[f'log_{c}'] = np.log1p(out[c])
    return out

Xtr_log = agregar_logs(Xtr_cap)
Xte_log = agregar_logs(Xte_cap)

print('Variables tras logs:', Xtr_log.shape[1])
print('¿Infinitos?', np.isinf(Xtr_log).sum().sum(), '(train) |', np.isinf(Xte_log).sum().sum(), '(test)')
````

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
sns.histplot(Xtr_log['TotalIncome'], ax=axes[0], kde=True, color='#e76f51')
axes[0].set_title(f'TotalIncome original (asimetría = {Xtr_log["TotalIncome"].skew():.2f})')
sns.histplot(Xtr_log['log_TotalIncome'], ax=axes[1], kde=True, color='#2a9d8f')
axes[1].set_title(f'log_TotalIncome (asimetría = {Xtr_log["log_TotalIncome"].skew():.2f})')
plt.suptitle('El logaritmo simetriza la distribución', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 6.7 Encoding de variables categóricas

> *"Cuando la variable categórica es **nominal**, se recomienda utilizar **one-hot encoding**."*

Convierte cada categoría en una columna binaria. `drop='first'` elimina una categoría de
referencia para evitar la **trampa de las variables dummy** (colinealidad perfecta: si conoces

### 🔴 El bug silencioso: `pd.get_dummies` por separado en train y test

`pd.get_dummies` genera columnas **según las categorías que encuentra en el DataFrame que
recibe**. Si en test falta una categoría que sí estaba en train (o aparece una nueva),
**las columnas no coinciden** y el modelo falla — o peor, no falla y predice basura porque
las columnas quedaron corridas.

Demostrémoslo:

````python
d_train = pd.get_dummies(Xtr_cat, drop_first=True)
d_test = pd.get_dummies(Xte_cat, drop_first=True)

print('Columnas generadas en TRAIN:', d_train.shape[1])
print('Columnas generadas en TEST :', d_test.shape[1])
print()
faltan = set(d_train.columns) - set(d_test.columns)
sobran = set(d_test.columns) - set(d_train.columns)
print('En TEST faltan columnas que TRAIN sí tiene :', faltan if faltan else '(ninguna, esta vez)')
print('En TEST sobran columnas que TRAIN no tiene :', sobran if sobran else '(ninguna, esta vez)')
print()
print('>> En ESTE dataset coinciden por suerte: todas las categorías aparecen en ambos.')
print('   Con una categoría rara (una zona con 3 clientes), test podría no tenerla y el')
print('   modelo recibiría una matriz con una columna menos → ValueError en producción,')
print('   o silenciosamente columnas desalineadas. NO SE DEBE DEPENDER DE LA SUERTE.')
````

### ✅ La forma correcta: `OneHotEncoder` con las categorías aprendidas en TRAIN

`handle_unknown='ignore'` hace que una categoría **nunca vista** en train se codifique como
todo-ceros en lugar de reventar. En producción esto es indispensable: siempre aparece un
valor nuevo que nadie anticipó.

````python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore')
ohe.fit(Xtr_cat)   # aprende las categorías SOLO de train

params_aprendidos['ohe'] = ohe

print('Categorías aprendidas en TRAIN:')
for col, cats in zip(cols_cat, ohe.categories_):
    print(f'  {col:18s} → {list(cats)}')
````

````python
nombres_ohe = ohe.get_feature_names_out(cols_cat)

Xtr_ohe = pd.DataFrame(ohe.transform(Xtr_cat), columns=nombres_ohe, index=Xtr_cat.index)
Xte_ohe = pd.DataFrame(ohe.transform(Xte_cat), columns=nombres_ohe, index=Xte_cat.index)  # transform

print('Dummies generadas:', Xtr_ohe.shape[1])
print('¿Train y test tienen EXACTAMENTE las mismas columnas, en el mismo orden?',
      list(Xtr_ohe.columns) == list(Xte_ohe.columns))
Xtr_ohe.head()
````

## 6.9 Escalamiento

> *"Hace que las variables que ingresan al modelo se encuentren en la misma escala.
> Se mantiene la distribución, pero con los valores transformados a una escala determinada."*

| Método | Fórmula | Resultado | Cuándo |
|---|---|---|---|
| **Estandarización normal** (Z-score) | $z = \dfrac{x - \mu}{\sigma}$ | media 0, desv. 1 | Datos ~normales; PCA; regresión regularizada |
| **Min-Max** | $x' = \dfrac{x - x_{min}}{x_{max} - x_{min}}$ | rango [0, 1] | Redes neuronales; cuando se requiere rango acotado |

**¿Cuándo es obligatorio escalar?**

- ✅ Modelos basados en **distancias**: KNN, K-Means, SVM, PCA
- ✅ Modelos con **regularización**: Ridge, Lasso, Logística con penalty
- ✅ Redes neuronales (convergencia del gradiente)
- ❌ **NO** lo necesitan: árboles, Random Forest, XGBoost (parten por umbrales, la escala les da igual)

> Del marco conceptual: *"siempre tiene que aplicarse en ML **no supervisado**"* — porque
> K-Means o PCA sin escalar quedan dominados por la variable de mayor magnitud. Si `ApplicantIncome`
> va en miles y `Dependents` en unidades, el PCA se convierte, literalmente, en un gráfico
> del ingreso.

````python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

scaler = StandardScaler()
scaler.fit(X_train_total)   # ⬅️ fit SOLO en train

params_aprendidos['scaler'] = scaler

X_train_scaled = pd.DataFrame(scaler.transform(X_train_total),
                              columns=X_train_total.columns, index=X_train_total.index)

# ✅ transform, NUNCA fit_transform. Este es EL error del notebook original.
X_test_scaled = pd.DataFrame(scaler.transform(X_test_total),
                             columns=X_test_total.columns, index=X_test_total.index)

print('X_train_scaled:', X_train_scaled.shape)
print('X_test_scaled :', X_test_scaled.shape)
print()
print('Train tras estandarizar → media ≈ 0, std ≈ 1 (por construcción):')
print(X_train_scaled[['ApplicantIncome', 'LoanAmount', 'TotalIncome']].describe().loc[['mean', 'std']].round(4))
print()
print('Test tras estandarizar → media y std NO son exactamente 0 y 1, ¡y eso está BIEN!:')
print(X_test_scaled[['ApplicantIncome', 'LoanAmount', 'TotalIncome']].describe().loc[['mean', 'std']].round(4))
print()
print('>> Que el test NO quede perfectamente centrado es la PRUEBA de que lo hicimos bien.')
print('   Si test tuviera media exactamente 0 y std exactamente 1, significaría que le hicimos')
print('   fit_transform: le habríamos dejado ver sus propios estadísticos. Ese "defecto"')
print('   es en realidad el certificado de que no hubo leakage.')
````

### 🔴 Cuantifiquemos el leakage del escalado

````python
scaler_malo = MinMaxScaler().fit(X_test_total)   # ❌ el error del notebook original
scaler_bueno = MinMaxScaler().fit(X_train_total)

comp = pd.DataFrame({
    'min_TRAIN': scaler_bueno.data_min_,
    'min_TEST': scaler_malo.data_min_,
    'max_TRAIN': scaler_bueno.data_max_,
    'max_TEST': scaler_malo.data_max_,
}, index=X_train_total.columns)
comp['max_difiere'] = (comp['max_TRAIN'] != comp['max_TEST'])

n_difieren = int(comp['max_difiere'].sum())
print(f'De {len(comp)} variables, {n_difieren} tendrían un MÁXIMO distinto si escaláramos con test.')
print()
print(comp[comp['max_difiere']].head(8).round(2))
print()
print('>> Consecuencia: un mismo cliente, con los mismos datos, recibiría un valor escalado')
print('   DISTINTO según el lote en el que llegue. El modelo se vuelve no determinista.')
print('   Un cliente aprobado el lunes podría ser rechazado el martes sin haber cambiado nada.')
````

## 6.11 Profundización: ¿imputar o no? ¿con qué?

El bloque anterior imputó numéricas con la mediana y categóricas con la moda. Pero la decisión
merece más rigor. Comparamos tres estrategias sobre las numéricas y vemos **cuánto cambian la
distribución** — porque imputar mal la deforma.

````python
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.experimental import enable_iterative_imputer  # noqa
from sklearn.impute import IterativeImputer

# Trabajamos sobre las numéricas ORIGINALES con nulos (antes de imputar)
num_cols = X_train.select_dtypes("number").columns.tolist()
con_nulos = [c for c in num_cols if X_train[c].isna().any()]
print("Numéricas con nulos:", con_nulos)

if con_nulos:
    c = con_nulos[0]
    orig = X_train[c].dropna()
    estrategias = {
        "Mediana (simple)": SimpleImputer(strategy="median"),
        "KNN (k=5)": KNNImputer(n_neighbors=5),
        "Iterativa (MICE)": IterativeImputer(random_state=seed, max_iter=10),
    }
    fig, axes = plt.subplots(1, len(estrategias) + 1, figsize=(4.2 * (len(estrategias) + 1), 3.6))
    sns.histplot(orig, kde=True, ax=axes[0], color=COLOR_NEUTRAL)
    axes[0].set_title(f"{c}\nORIGINAL (sin nulos)", fontsize=10)
    for j, (nombre, imp) in enumerate(estrategias.items(), start=1):
        vals = pd.DataFrame(imp.fit_transform(X_train[num_cols]), columns=num_cols)[c]
        sns.histplot(vals, kde=True, ax=axes[j], color=COLOR_BAD)
        axes[j].set_title(f"{nombre}\nμ={vals.mean():.0f} σ={vals.std():.0f}", fontsize=10)
    plt.suptitle("Efecto de la estrategia de imputación en la distribución", fontweight="bold")
    plt.tight_layout(); plt.show()
    print(">> La imputación simple concentra masa en un punto (pico artificial en la mediana).")
    print("   KNN e iterativa preservan mejor la forma, a costa de más cómputo y más supuestos.")
    print("   Regla: si el % de nulos es bajo, la mediana basta. Si es alto y hay correlación")
    print("   entre variables, KNN/iterativa valen la pena. SIEMPRE se ajustan solo en train.")
else:
    print("Tras la imputación del pipeline ya no quedan numéricas con nulos que comparar.")
````

### ¿Nulo o cero? La pregunta que decide la imputación

> *Si falta la **cantidad de transacciones** de un cliente, ¿qué imputo?* → casi siempre **cero**:
> el cliente no transó, por eso no aparece. Imputar la mediana le inventa actividad.

En nuestro caso, `CoapplicantIncome = 0` **no es un nulo**: es la ausencia de co-solicitante. Por
eso en 6.5 creamos `Tiene_Coaplicante` en vez de imputar. **Entender de dónde viene el dato es
más importante que la técnica.**

````python
pct_cero = (X_train["CoapplicantIncome"] == 0).mean()
pct_nulo = X_train["CoapplicantIncome"].isna().mean()
print(f"CoapplicantIncome — % de ceros: {pct_cero:.1%} | % de nulos: {pct_nulo:.1%}")
print(">> Los ceros NO son nulos: son 'no hay co-solicitante'. Son señal, no ausencia de dato.")
````

### Escalamiento: Standard vs MinMax vs Robust

El bloque usó `StandardScaler`. Comparémoslo con `RobustScaler` (usa mediana e IQR, resistente a
outliers) sobre una variable con cola.

````python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
c = "ApplicantIncome"
fig, axes = plt.subplots(1, 3, figsize=(14, 3.6))
for ax, (nombre, sc) in zip(axes, [("StandardScaler", StandardScaler()),
                                   ("MinMaxScaler", MinMaxScaler()),
                                   ("RobustScaler", RobustScaler())]):
    v = sc.fit_transform(X_train_total[[c]]).ravel()
    sns.histplot(v, kde=True, ax=ax, color=COLOR_NEUTRAL)
    ax.set_title(f"{nombre}\nrango [{v.min():.1f}, {v.max():.1f}]", fontsize=10)
    ax.set_xlabel("")
plt.suptitle("El escalado cambia la ESCALA, no la FORMA", fontweight="bold")
plt.tight_layout(); plt.show()
print(">> Las tres tienen la MISMA forma. Escalar no normaliza una distribución sesgada;")
print("   para eso está el log (6.6). RobustScaler es la mejor opción cuando quedan outliers.")
````
