````python
import warnings
warnings.filterwarnings("ignore")

from pipeline_io import (
    TARGET, SEGMENT_COL, SEED, load_stage, save_stage, save_report, save_table, load_report,
    set_plot_theme,
)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_selection import VarianceThreshold, chi2, f_classif, mutual_info_classif, RFE
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from statsmodels.stats.outliers_influence import variance_inflation_factor
from scipy import stats

set_plot_theme()

# --- Datos procesados (del NB06) ---
train_proc = load_stage("06_train_proc")
test_proc = load_stage("06_test_proc")
y_train = train_proc[TARGET]
y_test = test_proc[TARGET]
X_train_scaled = train_proc.drop(columns=[TARGET, SEGMENT_COL])
X_test_scaled = test_proc.drop(columns=[TARGET, SEGMENT_COL])

# --- Categóricas CRUDAS (del NB04), alineadas por índice para chi² ---
raw = load_stage("04_train")
X_train = raw.loc[X_train_scaled.index].drop(
    columns=[TARGET, SEGMENT_COL] + [c for c in ["maha_d2", "score_iso", "score_lof", "flag_outlier"] if c in raw])

# --- IV del NB05 ---
iv_table = load_report("iv_table")

# Alias para el código de esta sesión
seed = SEED
target = TARGET
y_tr = np.asarray(y_train)
# El bloque construye una versión MinMax [0,1] a partir de "X_train_total"; en la cadena
# modular esa matriz es la escalada del NB06.
X_train_total = X_train_scaled
X_test_total = X_test_scaled

print("X_train_scaled:", X_train_scaled.shape, "| X_train (crudo):", X_train.shape)
print("IV disponible para", len(iv_table["iv_por_variable"]), "variables crudas")


# ## 7.2 VarianceThreshold: el filtro automático de variables que no varían
#
# `VarianceThreshold` de sklearn elimina toda variable cuya **varianza** esté por debajo de un umbral.
# Es el filtro más barato que existe: **no mira el target**, sólo pregunta "¿esta columna se mueve?".
#
# Antes de aplicarlo, un experimento que deja una lección enorme. Nuestro `X_train_scaled` viene de un
# **StandardScaler**. ¿Qué varianza tiene cada columna?
````

````python
print('Varianza de las columnas de X_train_scaled (StandardScaler):\n')
print(X_train_scaled.var().describe().round(4).to_string())

sel_test = VarianceThreshold(threshold=0.01).fit(X_train_scaled)
print(f'\n¿Cuántas variables eliminaría VarianceThreshold(0.01) aquí? '
      f'{(~sel_test.get_support()).sum()} de {X_train_scaled.shape[1]}')
````

````python
# ¿Por qué la tabla de arriba dice 1.002 y no 1.000? No es ruido numérico: es ddof.
n_tr = len(X_train_scaled)
var_pandas = X_train_scaled.var().iloc[0]           # pandas -> ddof=1 (muestral)
var_numpy = X_train_scaled.var(ddof=0).iloc[0]      # numpy/sklearn -> ddof=0 (poblacional)

print('EL MISMO DATO, DOS ESTIMADORES DISTINTOS\n')
print(f'   pandas .var()          (ddof=1, muestral)    = {var_pandas:.6f}')
print(f'   .var(ddof=0)           (poblacional)         = {var_numpy:.6f}')
print(f'\n   El cociente entre ambos: {var_pandas / var_numpy:.6f}')
print(f'   Y n/(n-1) = {n_tr}/{n_tr - 1} = {n_tr / (n_tr - 1):.6f}   <- exactamente el mismo número')
print(f'\n-> StandardScaler divide por la desviación POBLACIONAL (ddof=0), así que en SU')
print(f'   escala la varianza es exactamente {var_numpy:.1f}. pandas la recalcula con ddof=1')
print(f'   y sale {var_pandas:.3f}. Ninguno está mal: miden cosas ligeramente distintas.')
````

### ⚠️ Todas las varianzas valen lo mismo (≈1). Y no es coincidencia.

**`StandardScaler` fuerza, por definición, media 0 y desviación 1 en cada columna.** Y si la desviación
es 1, la varianza es 1. **Todas. Siempre.**

> 🔎 **¿Por qué imprime 1.002 y no 1.000?** Detalle fino que vale oro, y la celda de arriba lo demuestra:
> `StandardScaler` divide por la desviación **poblacional** (`ddof=0`) y `pandas .var()` calcula la
> **muestral** (`ddof=1`). La diferencia es exactamente el factor `n/(n−1)`. No es un bug ni ruido
> numérico: son **dos estimadores distintos de la misma cantidad**.
>
> Llévate la pregunta, no el número: cuando dos librerías te dan resultados que difieren en el tercer
> decimal, lo correcto casi nunca es *"¿cuál está mal?"* — es **"¿qué está calculando cada una
> exactamente?"**. En un scorecard que se audita, esa pregunta te ahorra semanas.

> **Aplicar `VarianceThreshold` sobre datos estandarizados es literalmente inútil**: o no elimina nada
> (umbral < 1), o las elimina todas (umbral > 1). No hay punto intermedio.

Esto no es un bug del pipeline — `StandardScaler` es la elección correcta aquí (lo necesitan el PCA y la
regresión regularizada). Es una lección sobre la **varianza**:

> **La varianza tiene unidades y depende de la escala.** No es una propiedad intrínseca de la variable.
> "Esta variable tiene varianza baja" es una frase incompleta: **¿en qué escala?**

**La solución correcta:** para juzgar varianzas hay que ponerlas en una escala **común y acotada**, donde
las diferencias de dispersión sobrevivan. Eso es exactamente `MinMaxScaler` sobre [0, 1].

Usamos `X_train_total` (la matriz **sin escalar**) y le aplicamos un MinMax **local, sólo para este análisis**,
ajustado **únicamente sobre train** (nada de `fit` en test: eso sería *leakage*).

````python
# MinMax LOCAL, sólo para los análisis que necesitan una escala acotada y no negativa
# (VarianceThreshold y chi2). NO reemplaza al pipeline: X_train_scaled sigue siendo el oficial.
scaler_mm = MinMaxScaler()
scaler_mm.fit(X_train_total)            # ⬅️ fit SÓLO en train

X_train_mm = pd.DataFrame(scaler_mm.transform(X_train_total),
                          columns=X_train_total.columns, index=X_train_total.index)

print('X_train_mm — todo en [0, 1]:')
print(f'   mínimo global = {X_train_mm.values.min():.4f}')
print(f'   máximo global = {X_train_mm.values.max():.4f}')
print(f'   ¿algún valor negativo? {(X_train_mm.values < 0).any()}')
````

````python
# Varianza de cada variable en la escala MinMax [0, 1]
varianzas = X_train_mm.var().sort_values()

plt.figure(figsize=(11, 9))
colores = ['#c0392b' if v < 0.01 else '#2980b9' for v in varianzas.values]
plt.barh(varianzas.index, varianzas.values, color=colores)
plt.axvline(0.01, color='red', linestyle='--', linewidth=2, label='Umbral = 0.01')
plt.xlabel('Varianza (escala MinMax [0, 1])')
plt.title('Varianza por variable — ¿cuáles casi no se mueven?', fontsize=13, fontweight='bold')
plt.legend()
plt.tight_layout()
plt.show()
````

````python
UMBRAL_VAR = 0.01

selector_var = VarianceThreshold(threshold=UMBRAL_VAR)
selector_var.fit(X_train_mm)

mask = selector_var.get_support()
vars_conservadas = X_train_mm.columns[mask]
vars_eliminadas = X_train_mm.columns[~mask]

print(f'Umbral de varianza: {UMBRAL_VAR}  (sobre la escala MinMax)')
print(f'Variables conservadas: {mask.sum()} de {X_train_mm.shape[1]}')
print(f'\nELIMINADAS ({len(vars_eliminadas)}):')
for v in vars_eliminadas:
    s_raw = X_train_total[v]
    print(f'   - {v:<28} varianza = {X_train_mm[v].var():.6f} | '
          f'% de ceros = {(s_raw == 0).mean():.1%}')

# Interpretación de negocio de cada descarte, con los números de la propia tabla
print('\n\nPor qué cae cada una:')
for v in vars_eliminadas:
    s_raw = X_train_total[v]
    pct_no_cero = (s_raw != 0).mean()
    n_no_cero = int((s_raw != 0).sum())
    print(f'   - {v}: sólo el {pct_no_cero:.1%} de los clientes ({n_no_cero} de {len(s_raw)}) '
          f'tiene un valor distinto de 0,')
    print(f'     o sea que la columna es 0 en el {1 - pct_no_cero:.1%} de los casos. '
          f'Prácticamente no varía')
    print('     -> el modelo no tiene casos suficientes para aprender nada de ella.')
````

**Interpretación de negocio.** Cae `Married_missing`: es el indicador de "el estado civil venía nulo", y
como **casi ningún** cliente tenía ese campo vacío, la columna es 0 para la práctica totalidad de la
cartera (la celda de arriba imprime el porcentaje exacto). Prácticamente no varía → el modelo no tiene
casos suficientes para aprender nada de ella.

Ése es un descarte **razonable**. Pero ahora viene la parte importante.

### ⚠️ El peligro de subir un poquito el umbral

````python
# ¿Qué pasaría si fuéramos un poco más "exigentes"?
for umbral in [0.01, 0.02, 0.025]:
    elim = list(X_train_mm.columns[~VarianceThreshold(threshold=umbral)
                                   .fit(X_train_mm).get_support()])
    print(f'Umbral {umbral:<6} -> elimina {len(elim)}: {elim}')
````

Mira lo que pasa al subir el umbral a apenas **0.02**: se lleva por delante a **`%Cuota`** y a
**`Loan_Amount_Term`** — y al llegar a 0.025 la lista ya incluye a `ApplicantIncome` y `TotalIncome`.

Y `%Cuota` es **una de las variables que el Random Forest más usa** (lo verás en A.5.2: entra en el top 5 de
importancias). Tiene todo el sentido de negocio: **el nivel de endeudamiento sobre el ingreso** es
literalmente lo primero que mira un analista de riesgo.

Fíjate en la trampa completa: **`Loan_Amount_Term` es la misma variable del caso 4 de A.1**, la del CV más
bajo. Primero estuvimos a punto de descartarla por "no moverse", y ahora `VarianceThreshold` intenta
borrarla otra vez — por exactamente el mismo motivo, y con el mismo error de fondo.

¿Por qué su varianza es baja? Porque tiene **outliers extremos**: unos pocos clientes con ratios altísimos
estiran el rango a [0, 1] y aplastan al 99% de la población cerca de 0. **El problema no es la variable,
es el escalado.** MinMax es muy sensible a outliers.

> **Moraleja para la clase:** `VarianceThreshold` es un detector de *síntomas*, no un juez.
> **Varianza baja ≠ variable inútil.** Úsalo para que te avise **qué mirar**, nunca para borrar en automático.
> El caso extremo: un fraude ocurre en el 0.1% de los casos (varianza bajísima) y es exactamente lo que
> quieres predecir.

Por eso **no eliminamos nada todavía**. Seguimos con las 32 columnas y dejamos opinar a las técnicas que
**sí miran el target**.

## 7.3 Técnicas estadísticas: ahora sí miramos el target

El univariado no sabe nada de `Loan_Status`. Ahora preguntamos: **¿esta variable tiene relación con que el
cliente pague o no?** Cada técnica responde algo distinto:

| Técnica | Para qué tipo de variable | Qué mide | Detecta relaciones no lineales |
|---|---|---|---|
| **Chi-cuadrado** (`chi2`) | Categórica vs target categórico | ¿La distribución de la categoría cambia según el target? | Sí (es tabla de contingencia) |
| **ANOVA / F-test** (`f_classif`) | Numérica vs target categórico | ¿La media de la variable difiere entre buenos y malos? | **No** — sólo diferencia de medias |
| **Información mutua** (`mutual_info_classif`) | Cualquiera | Cuánta información comparte con el target | **Sí** — cualquier tipo de dependencia |

### 7.3.1 Chi-cuadrado sobre las categóricas crudas

La hipótesis nula del chi-cuadrado es **"la variable y el target son independientes"**.
Un **p-valor bajo (< 0.05)** nos permite rechazarla → *sí hay relación* → la variable sirve.

````python
# Usamos el X_train crudo para tener las categorías legibles ('Urban', 'Rural'...)
X_train_cat_raw = X_train.select_dtypes(['object','string'])

resultados_chi2 = []
for col in X_train_cat_raw.columns:
    # crosstab de la categoría contra el target (dropna elimina los nulos de esa columna)
    tabla = pd.crosstab(X_train_cat_raw[col].values, y_tr)
    chi2_stat, p_valor, gl, esperados = stats.chi2_contingency(tabla)

    # V de Cramér: tamaño del efecto, comparable entre variables (0 = nada, 1 = total)
    n_obs = tabla.values.sum()
    cramer_v = np.sqrt(chi2_stat / (n_obs * (min(tabla.shape) - 1)))

    resultados_chi2.append({
        'variable': col,
        'chi2': chi2_stat,
        'p_valor': p_valor,
        'grados_libertad': gl,
        'cramer_v': cramer_v,
        'significativa': 'Sí' if p_valor < 0.05 else 'No',
    })

chi2_df = pd.DataFrame(resultados_chi2).sort_values('chi2', ascending=False).reset_index(drop=True)
print('CHI-CUADRADO — categóricas crudas vs Loan_Status\n')
print(chi2_df.round(4).to_string(index=False))
````

Veamos una tabla de contingencia concreta para entender qué está midiendo el test.
Esto es lo que un analista de riesgo mira primero:

````python
tabla_pa = pd.crosstab(X_train['Property_Area'].values, y_tr, normalize='index')
tabla_pa.columns = ['Otorgado (0)', 'NO otorgado (1)']
print('Tasa de rechazo por zona de la propiedad:\n')
print((tabla_pa * 100).round(1).to_string())

plt.figure(figsize=(8, 5))
(tabla_pa['NO otorgado (1)'] * 100).sort_values().plot(kind='barh', color='#c0392b')
plt.axvline(y_tr.mean() * 100, color='black', linestyle='--', label=f'Tasa global ({y_tr.mean():.1%})')
plt.xlabel('% de créditos NO otorgados')
plt.title('Tasa de rechazo según zona de la propiedad', fontsize=13, fontweight='bold')
plt.legend()
plt.tight_layout()
plt.show()
````

### 7.3.2 Chi-cuadrado de sklearn + ANOVA F-test + Información mutua

Ahora las tres técnicas sobre toda la matriz. Pero antes, **un error que te vas a encontrar sí o sí**.

Intentemos correr `chi2` directamente sobre `X_train_scaled` (el que sale del `StandardScaler`):

````python
# Esto FALLA a propósito. Es el error más común al usar chi2.
try:
    chi2(X_train_scaled, y_tr)
    print('Funcionó (no debería).')
except ValueError as e:
    print('💥 ValueError:', e)
    print(f'\n   Mínimo de X_train_scaled = {X_train_scaled.values.min():.4f}')
    print('   -> StandardScaler centra en 0, así que la MITAD de los valores son negativos.')
````

### ¿Por qué `chi2` prohíbe los negativos?

No es un capricho de sklearn: es la **naturaleza del test**. El chi-cuadrado compara **frecuencias
observadas contra frecuencias esperadas** en una tabla de contingencia:

$$\chi^2 = \sum \frac{(observado - esperado)^2}{esperado}$$

Está pensado para **conteos**: "¿cuántos clientes de zona urbana fueron rechazados?". Un conteo **nunca
puede ser negativo** — no existen "−2 clientes". Cuando sklearn recibe una matriz numérica, trata cada
valor como si fuera una especie de *frecuencia o magnitud acumulable*. Un z-score de **−1.8 no significa
nada** en ese marco: no es un conteo, es "1.8 desviaciones por debajo del promedio".

**La solución:** llevar los datos a una escala **no negativa y acotada**. Ya tenemos exactamente eso:
`X_train_mm`, el MinMax [0, 1] que construimos en A.2.

Ojo con un matiz: **no** usamos `X_train_total` (la matriz cruda sin escalar) aunque tampoco tenga negativos.
¿Por qué? Porque el estadístico chi2 **crece con la magnitud** de la variable: `ApplicantIncome` (miles)
aplastaría a `Dependents` (0-3) sólo por estar medida en números más grandes, y el ranking sería una
mentira. MinMax pone a todas en el mismo rango → **los scores se vuelven comparables entre sí**.

````python
# --- Chi-cuadrado: necesita valores NO NEGATIVOS -> usamos la versión MinMax ---
chi2_vals, chi2_p = chi2(X_train_mm, y_tr)

# --- ANOVA F-test: ¿difiere la media entre clases? ---
# Funciona perfecto con negativos: internamente compara MEDIAS entre grupos y las
# diferencias de medias son invariantes al centrado. Usamos el escalado oficial.
f_vals, f_p = f_classif(X_train_scaled, y_tr)

# --- Información mutua: captura relaciones no lineales ---
# También acepta negativos: se basa en la DENSIDAD/vecindad de los valores, no en su signo.
mi_vals = mutual_info_classif(X_train_scaled, y_tr, random_state=seed)

ranking = pd.DataFrame({
    'variable': X_train_scaled.columns,
    'chi2': chi2_vals,
    'chi2_p': chi2_p,
    'F_anova': f_vals,
    'anova_p': f_p,
    'info_mutua': mi_vals,
})

# Rankings (1 = más importante) para poder compararlos entre técnicas
ranking['rank_chi2'] = ranking['chi2'].rank(ascending=False).astype(int)
ranking['rank_anova'] = ranking['F_anova'].rank(ascending=False).astype(int)
ranking['rank_mi'] = ranking['info_mutua'].rank(ascending=False).astype(int)
ranking['rank_promedio'] = ranking[['rank_chi2', 'rank_anova', 'rank_mi']].mean(axis=1)

ranking = ranking.sort_values('rank_promedio').reset_index(drop=True)

print('TABLA COMPARATIVA DE RANKINGS — 3 técnicas estadísticas\n')
print(ranking[['variable', 'chi2', 'chi2_p', 'F_anova', 'anova_p', 'info_mutua',
               'rank_chi2', 'rank_anova', 'rank_mi', 'rank_promedio']]
      .round(4).to_string(index=False))
````

````python
# ¿EN QUÉ COINCIDEN Y EN QUÉ NO? Que lo conteste la tabla, no nosotros.
cols_r = ['rank_chi2', 'rank_anova', 'rank_mi']
TOP = 5

tops = {c: set(ranking.nsmallest(TOP, c)['variable']) for c in cols_r}
consenso = tops['rank_chi2'] & tops['rank_anova'] & tops['rank_mi']

print(f'CONSENSO — variables en el top {TOP} de LAS TRES técnicas ({len(consenso)}):\n')
for v in ranking[ranking['variable'].isin(consenso)]['variable']:
    r = ranking[ranking['variable'] == v].iloc[0]
    print(f'   {v:<34} chi2 #{r["rank_chi2"]:<3} ANOVA #{r["rank_anova"]:<3} MI #{r["rank_mi"]}')
print('\n-> Cuando las tres coinciden Y con margen, la señal es sólida.')

# Los desacuerdos: variables donde el ranking cambia MUCHO según la técnica
ranking['desacuerdo'] = ranking[cols_r].max(axis=1) - ranking[cols_r].min(axis=1)
print(f'\n\nDESACUERDO — donde las técnicas más se contradicen (top 5):\n')
print(ranking.nlargest(5, 'desacuerdo')[['variable'] + cols_r + ['desacuerdo']]
      .to_string(index=False))
print('\n-> ANOVA sólo compara MEDIAS; MI ve CUALQUIER dependencia. Una variable alta en')
print('   MI y baja en ANOVA es señal de una relación NO LINEAL (y viceversa).')

# El aviso de magnitud: rankear no es lo mismo que haber encontrado algo
print('\n\n⚠️  OJO CON LA MAGNITUD, no sólo con la posición:\n')
mi_top = ranking.nsmallest(TOP, 'rank_mi')
mi_max = ranking['info_mutua'].max()
for _, r in mi_top.iterrows():
    marca = '  <- ¡minúsculo!' if r['info_mutua'] < mi_max / 4 else ''
    print(f'   MI #{r["rank_mi"]:<3} {r["variable"]:<34} info_mutua = {r["info_mutua"]:.5f}{marca}')
n_mi_cero = int((ranking['info_mutua'] <= 1e-12).sum())
print(f'\n-> {n_mi_cero} de {len(ranking)} variables tienen información mutua = 0.')
print('   Un ranking SIEMPRE ordena algo, aunque no haya nada que ordenar.')
````

Un gráfico vale más que la tabla. Comparamos los tres rankings lado a lado:

````python
fig, axes = plt.subplots(1, 3, figsize=(18, 7))

configs = [
    ('chi2', 'Chi-cuadrado', '#2980b9'),
    ('F_anova', 'ANOVA (F-test)', '#27ae60'),
    ('info_mutua', 'Información mutua', '#8e44ad'),
]

for ax, (col, titulo, color) in zip(axes, configs):
    datos = ranking.sort_values(col, ascending=True)
    ax.barh(datos['variable'], datos[col], color=color)
    ax.set_title(titulo, fontsize=13, fontweight='bold')
    ax.set_xlabel('Score')

axes[0].set_ylabel('Variable')
plt.suptitle('¿Coinciden las técnicas sobre qué variables importan?', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
````

**Lo que debes observar** (los números concretos los imprime la celda de arriba; aquí va la lectura):

- Las dummies de **`Credit_History`** (`Sin_historial_OK` e `Historial_OK`) ganan **por goleada en las tres
  técnicas**, y por un margen enorme. Cuando las tres coinciden **y con esa distancia**, la señal es sólida.
  Tiene todo el sentido de negocio: el mejor predictor de si alguien va a pagar es **si ya pagó antes**.
- **Fuera del podio, las técnicas dejan de coincidir** — y ahí está lo interesante. Mira la tabla de
  *desacuerdo*: hay variables que una técnica pone en el top y otra manda al fondo. **No es que una se
  equivoque: es que preguntan cosas distintas.**
  - **ANOVA** sólo sabe comparar **medias** entre buenos y malos. Si una variable separa las clases sin que
    su media cambie (por ejemplo, con una relación en forma de U), ANOVA **no la ve**.
  - **Información mutua** detecta **cualquier** dependencia, lineal o no. Una variable que MI rankea alto y
    ANOVA bajo es tu candidata a **relación no lineal**.
  - **Chi-cuadrado** trabaja sobre frecuencias, así que brilla con las dummies y es más torpe con las
    continuas.
- **⚠️ Mira la MAGNITUD, no sólo la posición.** Es el aviso más importante de esta celda. El estimador de
  información mutua usa vecinos más cercanos y es **ruidoso con variables casi constantes**: puede colocar una
  variable `*_missing` en un puesto alto con un valor de MI **minúsculo**. Que algo sea "el #4" no significa
  nada si el #1 vale cinco veces más. **Un ranking siempre ordena algo, aunque no haya nada que ordenar.**
- La información mutua da **0** a varias variables. No significa "no sirve", significa "no aporta información
  *por sí sola*". Podría aportar **en combinación** con otra. Los filtros univariados son ciegos a las
  interacciones — ésa es su gran limitación, y es la que arreglan los métodos wrapper de A.5.

## 7.4 Matriz de correlación y multicolinealidad

Hasta aquí miramos cada variable **contra el target**. Ahora las miramos **entre ellas**.

El problema se llama **multicolinealidad**: dos o más variables que dicen prácticamente lo mismo.
Nosotros mismos la creamos al hacer feature engineering:

```
TotalIncome = ApplicantIncome + CoapplicantIncome     <- ¡combinación lineal exacta!
log_TotalIncome = log(TotalIncome)                    <- la misma info, otra escala
%Cuota = Cuota / TotalIncome                          <- comparte info con ambas
```

````python
corr = X_train_scaled.corr()

# Con 32 variables, anotar cada celda hace el gráfico ilegible: usamos sólo color.
plt.figure(figsize=(15, 12.5))
mask_sup = np.triu(np.ones_like(corr, dtype=bool))  # ocultamos el triángulo superior (es espejo)
sns.heatmap(corr, mask=mask_sup, annot=False, cmap='RdBu_r',
            center=0, vmin=-1, vmax=1, square=True,
            linewidths=0.4, cbar_kws={'label': 'Correlación de Pearson', 'shrink': 0.7})
plt.title('Matriz de correlación de Pearson — X_train_scaled (32 variables)',
          fontsize=14, fontweight='bold', pad=15)
plt.xticks(fontsize=9)
plt.yticks(fontsize=9)
plt.tight_layout()
plt.show()
````

Busca los **cuadros rojos oscuros fuera de la diagonal**: ésos son los pares problemáticos.
Se distingue a simple vista un "bloque" de variables de ingreso/monto que se parecen entre sí.

Con 32 variables el heatmap sirve para **ver el patrón**, no para leer números. Para eso, la lista explícita:

**Pearson vs Spearman** — vale la pena verlo, porque nuestras variables de ingreso tienen colas muy largas:

- **Pearson** mide relación **lineal**. Es sensible a outliers.
- **Spearman** mide relación **monótona** (usa rangos, no valores). Es robusto a outliers y a colas largas.

Si Spearman >> Pearson, hay una relación fuerte pero *no lineal*.

````python
pares = []
cols = X_train_scaled.columns

for i in range(len(cols)):
    for j in range(i + 1, len(cols)):
        a, b = cols[i], cols[j]
        r_p = X_train_scaled[a].corr(X_train_scaled[b], method='pearson')
        r_s = X_train_scaled[a].corr(X_train_scaled[b], method='spearman')
        pares.append({'var_1': a, 'var_2': b, 'pearson': r_p, 'spearman': r_s,
                      'abs_pearson': abs(r_p)})

pares_df = pd.DataFrame(pares).sort_values('abs_pearson', ascending=False)

UMBRAL_CORR = 0.8
pares_altos = pares_df[pares_df['abs_pearson'] > UMBRAL_CORR]

print(f'PARES CON |r| > {UMBRAL_CORR} — candidatos a multicolinealidad\n')
if len(pares_altos):
    print(pares_altos[['var_1', 'var_2', 'pearson', 'spearman']].round(3).to_string(index=False))
else:
    print('No hay pares por encima del umbral.')

print(f'\n\nTop 10 pares más correlacionados (para ver Pearson vs Spearman):\n')
print(pares_df.head(10)[['var_1', 'var_2', 'pearson', 'spearman']].round(3).to_string(index=False))
````

````python
# ¿De dónde salen esos pares? Casi todos los creamos nosotros. Que lo diga el dato.
# Las variables que venían en el CSV original (todo lo demás lo inventamos en el
# feature engineering: logaritmos, ratios, sumas, banderas...).
vars_originales = {'Dependents', 'ApplicantIncome', 'CoapplicantIncome',
                   'LoanAmount', 'Loan_Amount_Term'}


def _origen_par(a, b):
    """Clasifica de dónde viene la redundancia de un par."""
    if f'log_{a}' == b or f'log_{b}' == a:
        return 'variable vs su propio log'
    derivadas = [v for v in (a, b) if v not in vars_originales]
    if len(derivadas) == 2:
        return 'dos variables derivadas'
    if len(derivadas) == 1:
        return 'derivada vs su ingrediente'
    return 'venía así en el CSV'


pares_altos = pares_altos.copy()
pares_altos['origen'] = [_origen_par(a, b)
                         for a, b in zip(pares_altos['var_1'], pares_altos['var_2'])]

n_autoinfligidos = int((pares_altos['origen'] != 'venía así en el CSV').sum())

print(f'DE DÓNDE SALEN LOS {len(pares_altos)} PARES CON |r| > {UMBRAL_CORR}\n')
print(pares_altos[['var_1', 'var_2', 'pearson', 'origen']].round(3).to_string(index=False))

n_del_csv = len(pares_altos) - n_autoinfligidos
print(f'\n-> {n_autoinfligidos} de {len(pares_altos)} pares involucran al menos UNA variable que')
print('   NO venía en el CSV: la inventamos nosotros en el feature engineering.')
if n_del_csv == 0:
    print('   NINGUNO surge de dos variables originales del dataset. Ni uno solo.')
else:
    print(f'   Sólo {n_del_csv} surge(n) de dos variables originales del dataset.')
print('\n   Esto NO es un descubrimiento sobre los clientes: es el eco de nuestras propias')
print('   decisiones. Creamos las redundancias hace tres secciones y las estamos')
print('   redescubriendo ahora, con una herramienta estadística, como si fueran un hallazgo.')
````

**Resultado revelador.** La correlación encuentra un puñado de pares por encima de 0.8, y el veredicto de
la celda anterior es demoledor: **prácticamente todos involucran alguna variable que inventamos nosotros**
en el feature engineering — logaritmos, ratios, sumas, banderas. No es un hallazgo sobre los clientes: es
**el eco de nuestras propias decisiones**, que redescubrimos tres secciones después con una herramienta
estadística y cierta cara de sorpresa.

El par más extremo de la lista tiene una explicación preciosa: `log_CoapplicantIncome` vale ~0 cuando no hay
co-solicitante y algo grande cuando sí lo hay. O sea, **casi es** la variable binaria `Tiene_Coaplicante`.
Dos columnas, una sola idea.

**Pero ahora viene la pregunta que de verdad importa**, y no es "¿qué encontró la correlación?" sino
**"¿qué es capaz de decirnos cuando la encuentra?"**.

Nosotros construimos `TotalIncome` **exactamente** como `ApplicantIncome + CoapplicantIncome`: no es que se
parezcan, es que **una ES la suma de las otras dos**. Esa columna es 100% redundante — sobra, sin discusión.
La pregunta es si la matriz de correlación es capaz de **decirnos eso**. Vamos a verlo de cerca.

### VIF (Variance Inflation Factor): el detector serio de multicolinealidad

La matriz de correlación sólo ve **pares**. Pero la multicolinealidad puede ser **múltiple**: la variable A puede
ser predecible casi perfectamente a partir de B + C + D, **sin estar muy correlacionada con ninguna
individualmente**. Es justo nuestro caso.

El VIF resuelve eso. Para cada variable, hace una regresión de esa variable **contra todas las demás**:

$$VIF_i = \frac{1}{1 - R^2_i}$$

donde $R^2_i$ es qué tan bien las **otras** variables predicen a la variable $i$.

| VIF | Lectura |
|---|---|
| ≈ 1 | La variable es independiente. Perfecto. |
| 1 - 5 | Correlación moderada. Aceptable. |
| 5 - 10 | Problemático. Revisar. |
| > 10 | **Multicolinealidad severa.** Hay que actuar. |
| ∞ | **Redundancia exacta**: la variable es una combinación lineal perfecta de otras. |

````python
def calcular_vif(df):
    """VIF de cada variable. Requiere agregar una constante (el intercepto)."""
    X_vif = df.copy()
    X_vif.insert(0, 'const', 1.0)  # statsmodels necesita el intercepto explícito

    filas = []
    for i, col in enumerate(X_vif.columns):
        if col == 'const':
            continue
        filas.append({'variable': col, 'VIF': variance_inflation_factor(X_vif.values, i)})

    vif_df = pd.DataFrame(filas).sort_values('VIF', ascending=False).reset_index(drop=True)

    def _sem(v):
        if not np.isfinite(v):
            return 'REDUNDANCIA EXACTA'
        if v > 10:
            return 'severa'
        if v > 5:
            return 'problemática'
        if v > 1.5:
            return 'moderada'
        return 'OK'

    vif_df['diagnostico'] = vif_df['VIF'].apply(_sem)
    return vif_df


vif_resultado = calcular_vif(X_train_scaled)
print('VIF — X_train_scaled\n')
print(vif_resultado.round(2).to_string(index=False))
````

````python
# La comparación que importa: ¿qué vio cada herramienta?
r_ti_ai = X_train_scaled['TotalIncome'].corr(X_train_scaled['ApplicantIncome'])
r_ti_ci = X_train_scaled['TotalIncome'].corr(X_train_scaled['CoapplicantIncome'])
vif_ti = vif_resultado.loc[vif_resultado['variable'] == 'TotalIncome', 'VIF'].values[0]

# El VIF responde "¿qué % de esta variable puedo reconstruir con TODAS las demás?": VIF = 1/(1-R²)
r2_ti = 1 - 1 / vif_ti

# Nada de veredictos escritos a mano: que los decida el dato.
def _veredicto_corr(r):
    return 'POR ENCIMA' if abs(r) > UMBRAL_CORR else 'por debajo'

print('EL CASO "TotalIncome" — correlación vs VIF\n')
print('   Sabemos, porque la construimos nosotros, que:')
print('       TotalIncome = ApplicantIncome + CoapplicantIncome')
print('   O sea: esta columna es 100% redundante. Veamos qué dice cada herramienta.\n')

print('   MIRANDO DE A PARES (matriz de correlación):')
print(f'      corr(TotalIncome, ApplicantIncome)   = {r_ti_ai:.3f} '
      f' -> {_veredicto_corr(r_ti_ai)} del umbral {UMBRAL_CORR}')
print(f'      corr(TotalIncome, CoapplicantIncome) = {r_ti_ci:.3f} '
      f' -> {_veredicto_corr(r_ti_ci)} del umbral {UMBRAL_CORR}')

print('\n   MIRANDO CONTRA TODAS A LA VEZ (VIF):')
print(f'      VIF(TotalIncome) = {vif_ti:,.1f}  ->  R² = {r2_ti:.4f}')
print(f'      Traducción: las demás variables reconstruyen TotalIncome al {r2_ti:.1%}.')

print('\n' + '   ' + '-' * 66)
print('   LA LECCIÓN:\n')
print(f'   Ninguna correlación INDIVIDUAL dice "esta columna sobra". Contra')
print(f'   CoapplicantIncome da {r_ti_ci:.3f}: prácticamente nada, no se parecen. Y contra')
print(f'   ApplicantIncome da {r_ti_ai:.3f}, que como mucho sugiere "se parecen bastante"')
print('   -> redundancia PARCIAL, la clase de cosa que uno deja pasar todos los días.')
print('\n   Pero TotalIncome no "se parece" a nada: ES la suma exacta de las otras dos.')
print(f'   Eso NO se ve en ninguna correlación por pares. Sólo lo ve el VIF, que pregunta')
print('   "¿puedo reconstruir esta variable con TODAS las demás juntas?" — y responde')
print(f'   que sí, al {r2_ti:.1%}.')
print('\n   Ésa es exactamente la diferencia entre MIRAR DE A PARES y MIRAR EN CONJUNTO.')
print('   ' + '-' * 66)
````

**Éste es el punto central de toda la sección.**

`TotalIncome` nació de `ApplicantIncome + CoapplicantIncome`. Es una identidad, no un parecido: esa columna
**sobra**, y lo sabemos con certeza porque la escribimos nosotros. Y sin embargo, mira lo que dice cada
herramienta:

| Herramienta | Veredicto |
|---|---|
| Correlación con `CoapplicantIncome` | ✅ *"todo bien, no se parecen en nada"* |
| Correlación con `ApplicantIncome` | ⚠️ *"alta, pero es **una** variable: parece redundancia parcial"* |
| **VIF** | 🚨 **"multicolinealidad severa: es reconstruible casi al 100%"** |

**Ninguna correlación individual dice "esta columna sobra".** La más alta de las dos sugiere, como mucho,
*"estas dos se parecen bastante"* — un diagnóstico que un analista archiva sin más, porque variables que se
parecen bastante hay a montones y no se borran por eso. Y la otra, directamente, no ve nada.

El VIF hace la pregunta correcta: *"¿puedo reconstruir esta variable usando **todas** las demás **a la
vez**?"*. Y la respuesta es: **casi perfectamente**.

> 🎓 **La moraleja, y es la que se llevan a casa:** la matriz de correlación **no es una versión barata del
> VIF**. Es una herramienta que responde **otra pregunta**. Puede darte 32×32 números, todos por debajo de
> tu umbral, y aun así tener una variable que sobra al 100% delante de las narices. La redundancia **no
> tiene por qué vivir en un par**: puede estar repartida entre tres, cinco o diez columnas, y entonces
> **ningún par la delata**.

*(Nota: si el pipeline no capeara los outliers de `ApplicantIncome`, la relación sería una identidad exacta
y el VIF saldría literalmente `inf`. El capeo rompe la igualdad exacta y deja un VIF alto pero finito.)*

Y `TotalIncome` **ni siquiera es el peor caso de la tabla**: por encima de ella están los logaritmos y los
ratios (`log_CoapplicantIncome`, `log_Cuota`, `log_LoanAmount`, `Tiene_Coaplicante`...), todos
**redundancias que nos autoinfligimos** al crear versiones alternativas de variables que ya teníamos.

### ¿Por qué esto importa TANTO en regresión logística?

La regresión logística estima un coeficiente $\beta_i$ por variable, y ese coeficiente significa:
*"cuánto cambia el log-odds si muevo esta variable **manteniendo las demás constantes**"*.

**Si `TotalIncome ≈ ApplicantIncome + CoapplicantIncome`, esa frase pierde sentido**: es (casi) imposible mover
`ApplicantIncome` manteniendo `TotalIncome` constante — al subir una, sube la otra. El modelo no sabe a cuál
de las tres columnas atribuirle el efecto del ingreso, y entonces:

1. **Los coeficientes se vuelven inestables.** Cambias 5 filas del train y un β pasa de +2.0 a −1.5.
2. **Los signos se invierten sin sentido.** Aparece "a mayor ingreso, mayor riesgo" — y no puedes explicárselo al regulador.
3. **Los errores estándar explotan** → los p-valores dejan de ser confiables → concluyes que una variable buena "no es significativa".
4. **El poder predictivo NO se cae.** Ésta es la trampa: el AUC sigue igual. La multicolinealidad daña la
   **interpretabilidad**, no necesariamente el accuracy.

> **Por eso en banca importa tanto**: un modelo de riesgo no sólo predice, tiene que **justificar**. Si el coeficiente
> de ingreso sale negativo, el modelo no pasa validación interna ni auditoría regulatoria — aunque prediga bien.
>
> Los modelos de árboles (Random Forest, XGBoost) **no sufren** de esto para predecir: no estiman coeficientes.
> Pero sí les afecta la *importancia* de variables, que se reparte arbitrariamente entre las redundantes.

````python
# Demostración: quitemos las redundancias que creamos nosotros mismos.
# Criterio: nos quedamos con UNA variable por "idea de negocio".
#   - los log_* duplican a su variable original -> fuera
#   - TotalIncome se deduce de Applicant + Coapplicant -> fuera
#   - Tiene_Coaplicante ya está contenida en CoapplicantIncome -> fuera
vars_redundantes = ['log_ApplicantIncome', 'log_CoapplicantIncome', 'log_LoanAmount',
                    'log_TotalIncome', 'log_Cuota', 'TotalIncome', 'Tiene_Coaplicante']
vars_redundantes = [v for v in vars_redundantes if v in X_train_scaled.columns]

X_sin_redundantes = X_train_scaled.drop(columns=vars_redundantes)
vif_limpio = calcular_vif(X_sin_redundantes)

print(f'Variables eliminadas ({len(vars_redundantes)}): {vars_redundantes}\n')
print('VIF DESPUÉS de eliminar redundancias:\n')
print(vif_limpio.round(2).to_string(index=False))

max_antes = vif_resultado['VIF'].replace([np.inf, -np.inf], np.nan).max()
max_despues = vif_limpio['VIF'].replace([np.inf, -np.inf], np.nan).max()
print(f'\n\nVIF máximo ANTES  : {max_antes:,.1f}')
print(f'VIF máximo DESPUÉS: {max_despues:,.1f}')
print(f'Variables con VIF > 10 antes  : {(vif_resultado["VIF"] > 10).sum()}')
print(f'Variables con VIF > 10 después: {(vif_limpio["VIF"] > 10).sum()}')
````

Quitando 7 columnas que **no aportaban información nueva**, el VIF máximo se desploma.

Y aquí está lo importante para el negocio: **no perdimos nada**. `log_ApplicantIncome` y `ApplicantIncome`
contienen exactamente la misma información sobre el cliente — sólo que en dos escalas distintas. Tener las
dos no hace al modelo más inteligente; lo hace más **inestable** y más difícil de explicar.

> **Regla práctica de banca:** una idea de negocio → una variable. Si tienes "ingreso", "log del ingreso",
> "ingreso total" e "ingreso por persona", elige **la que mejor se explique en un comité de riesgos** y
> descarta el resto.

## 7.5 Métodos wrapper: que el modelo elija

Los filtros anteriores (chi2, ANOVA, MI) evalúan las variables **una por una, sin modelo**. Rápidos, pero ciegos
a las interacciones.

Los métodos **wrapper** entrenan el modelo de verdad y le preguntan a él. Son más lentos pero capturan
combinaciones.

### 7.5.1 RFE — Recursive Feature Elimination

La idea es de eliminación por rondas:
1. Entrena el modelo con **todas** las variables.
2. Mira la importancia de cada una y **elimina la peor**.
3. **Reentrena** con las que quedan y repite.

El paso 3 es la clave: al reentrenar, las importancias se recalculan. Una variable que parecía mala
puede volverse útil cuando desaparece su redundante.

````python
# RFE con regresión logística
modelo_rfe = LogisticRegression(max_iter=2000, random_state=seed)
rfe = RFE(estimator=modelo_rfe, n_features_to_select=8, step=1)
rfe.fit(X_train_scaled, y_tr)

rfe_df = pd.DataFrame({
    'variable': X_train_scaled.columns,
    'seleccionada': rfe.support_,
    'orden_eliminacion': rfe.ranking_,  # 1 = seleccionada; mayor = eliminada antes
}).sort_values('orden_eliminacion').reset_index(drop=True)

print('RFE — Recursive Feature Elimination (LogisticRegression, top 8)\n')
print(rfe_df.to_string(index=False))
print(f'\nSeleccionadas: {list(X_train_scaled.columns[rfe.support_])}')
````

### 7.5.2 Coeficientes de la regresión logística vs importancias del Random Forest

Dos modelos, dos formas de decir "esta variable importa":

- **Regresión logística** → el **coeficiente** β. Tiene **signo**: dice *en qué dirección* empuja la variable.
  Como los datos están escalados, los coeficientes son comparables entre sí.
- **Random Forest** → `feature_importances_`. Mide cuánto reduce la impureza esa variable. **No tiene signo**:
  dice "importa", no "importa en qué dirección".

````python
# --- Regresión logística ---
log_reg = LogisticRegression(max_iter=2000, random_state=seed)
log_reg.fit(X_train_scaled, y_tr)

coefs = pd.DataFrame({
    'variable': X_train_scaled.columns,
    'coeficiente': log_reg.coef_[0],
    'abs_coef': np.abs(log_reg.coef_[0]),
}).sort_values('abs_coef', ascending=False).reset_index(drop=True)

# --- Random Forest ---
rf = RandomForestClassifier(n_estimators=300, random_state=seed, n_jobs=-1)
rf.fit(X_train_scaled, y_tr)

importancias = pd.DataFrame({
    'variable': X_train_scaled.columns,
    'importancia': rf.feature_importances_,
}).sort_values('importancia', ascending=False).reset_index(drop=True)

print('COEFICIENTES — Regresión logística (signo = dirección del efecto)\n')
print(coefs.round(4).to_string(index=False))
print('\n\nIMPORTANCIAS — Random Forest\n')
print(importancias.round(4).to_string(index=False))
````

````python
fig, axes = plt.subplots(1, 2, figsize=(17, 7))

# Coeficientes con color según signo
c_plot = coefs.sort_values('coeficiente')
colores_coef = ['#c0392b' if c > 0 else '#27ae60' for c in c_plot['coeficiente']]
axes[0].barh(c_plot['variable'], c_plot['coeficiente'], color=colores_coef)
axes[0].axvline(0, color='black', linewidth=1)
axes[0].set_title('Regresión logística: coeficientes\n(rojo = aumenta riesgo de NO otorgar)',
                  fontsize=12, fontweight='bold')
axes[0].set_xlabel('Coeficiente β')

i_plot = importancias.sort_values('importancia')
axes[1].barh(i_plot['variable'], i_plot['importancia'], color='#2980b9')
axes[1].set_title('Random Forest: feature_importances_\n(sin signo, sólo magnitud)',
                  fontsize=12, fontweight='bold')
axes[1].set_xlabel('Importancia')

plt.tight_layout()
plt.show()
````

## 7.6 Consolidando: ¿en qué coinciden todos los métodos?

Ningún método es la verdad. La decisión real se toma **cruzando los rankings** y agregando criterio de negocio.

````python
consolidado = pd.DataFrame({'variable': X_train_scaled.columns})

consolidado = (consolidado
               .merge(ranking[['variable', 'rank_chi2', 'rank_anova', 'rank_mi']], on='variable')
               .merge(rfe_df[['variable', 'seleccionada']].rename(
                   columns={'seleccionada': 'rfe_selecciona'}), on='variable'))

consolidado['rank_coef_logreg'] = consolidado['variable'].map(
    coefs.set_index('variable')['abs_coef'].rank(ascending=False)).astype(int)
consolidado['rank_rf'] = consolidado['variable'].map(
    importancias.set_index('variable')['importancia'].rank(ascending=False)).astype(int)

cols_rank = ['rank_chi2', 'rank_anova', 'rank_mi', 'rank_coef_logreg', 'rank_rf']
consolidado['rank_final'] = consolidado[cols_rank].mean(axis=1)
consolidado = consolidado.sort_values('rank_final').reset_index(drop=True)

print('RANKING CONSOLIDADO — 5 métodos de selección\n')
print(consolidado.round(2).to_string(index=False))

print('\n\nTOP 8 por consenso:')
for i, v in enumerate(consolidado['variable'].head(8), 1):
    print(f'   {i}. {v}')
````

## 7.6 Consenso final: correlación/asociación vs IV vs multivariada

Juntamos todas las señales en una sola tabla y decidimos. Una variable **entra** si aporta señal
(ANOVA/MI/IV) y **no** es redundante (VIF bajo). El IV viene del NB05 (sobre variables crudas);
lo mapeamos a las features derivadas por su variable de origen cuando aplica.

````python
# Señal univariada sobre las features escaladas
f_scores, f_p = f_classif(X_train_scaled, y_tr)
mi = mutual_info_classif(X_train_scaled, y_tr, random_state=seed)

consenso = pd.DataFrame({
    "variable": X_train_scaled.columns,
    "anova_F": np.round(f_scores, 2),
    "anova_signif": f_p < 0.05,
    "info_mutua": np.round(mi, 4),
})

# VIF (multicolinealidad) — reusamos calcular_vif definido arriba
vif_df = calcular_vif(X_train_scaled).set_index("variable")["VIF"]
consenso["VIF"] = consenso["variable"].map(vif_df).round(2)
consenso["multicolineal"] = consenso["VIF"] > 10

# IV: mapeamos por prefijo de variable cruda (ApplicantIncome -> log_ApplicantIncome, etc.)
iv_map = iv_table["iv_por_variable"]
def iv_de(col):
    for base, iv in iv_map.items():
        if base in col:
            return iv
    return np.nan
consenso["IV_origen"] = consenso["variable"].map(iv_de).round(4)

consenso = consenso.sort_values("info_mutua", ascending=False).reset_index(drop=True)
print(consenso.to_string(index=False))
````

### Decisión

Regla transparente: **descarta** si (no aporta señal: no significativa en ANOVA y MI≈0) **o**
(es redundante: VIF > 10 y existe otra que aporta lo mismo). Todo lo demás **entra**.

````python
descartar = consenso[
    ((~consenso["anova_signif"]) & (consenso["info_mutua"] < 0.001))
    | (consenso["VIF"] > 30)
]["variable"].tolist()

seleccionadas = [c for c in X_train_scaled.columns if c not in descartar]

print("=" * 60)
print("SELECCIÓN DE VARIABLES")
print("=" * 60)
print(f"  De {X_train_scaled.shape[1]} features → {len(seleccionadas)} seleccionadas")
print(f"  Descartadas ({len(descartar)}): {descartar if descartar else 'ninguna'}")
print("=" * 60)
print()
print(">> Nota honesta: en este dataset casi todo aporta algo y la multicolinealidad severa")
print("   viene de features que NOSOTROS creamos (TotalIncome = suma, log_X = transformación).")
print("   El VIF las delata; el criterio decide con qué versión quedarse.")
````

## 7.7 Exportación

````python
train_sel = train_proc[seleccionadas + [SEGMENT_COL, TARGET]]
test_sel = test_proc[seleccionadas + [SEGMENT_COL, TARGET]]

save_stage(train_sel, "07_train_sel")
save_stage(test_sel, "07_test_sel")

reporte = {
    "seleccionadas": seleccionadas,
    "descartadas": descartar,
    "n_antes": int(X_train_scaled.shape[1]),
    "n_despues": len(seleccionadas),
    "consenso": consenso.to_dict(orient="records"),
}
save_report(reporte, "seleccion")
save_table(consenso.set_index("variable"), "seleccion_consenso")

print(f"07_train_sel → {train_sel.shape} | 07_test_sel → {test_sel.shape} | seleccion_consenso.csv")
````
