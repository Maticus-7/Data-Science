# 1. Configuración del Entorno y Carga de Datos

```python
# ==========================================
# 1. CONFIGURACIÓN DEL ENTORNO Y LIBRERÍAS
# ==========================================
import warnings
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Librerías estadísticas y de Machine Learning
from scipy import stats
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Configuración visual para gráficos limpios
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11
warnings.filterwarnings('ignore')

print("✅ Librerías importadas correctamente.")

# ==========================================
# 2. CARGA DE METADATA Y DATASET
# ==========================================
# 2.1 Cargar Metadata (para entender el contexto de las variables)
try:
    with open('metadata.json', 'r', encoding='utf-8') as f:
        metadata = json.load(f)
    print("\n📘 Metadata cargada con éxito. Información general:")
    print(json.dumps(metadata, indent=2, ensure_ascii=False)[:500] + "...\n")
except Exception as e:
    print(f"⚠️ No se pudo cargar 'metadata.json': {e}")

# 2.2 Carga robusta del Dataset manejando nulos ocultos
# Es común en datasets de internet que los nulos estén como '?', 'NA', '--' o 'None'
valores_nulos_ocultos = ['?', 'NA', '--', 'None', 'null', '', ' ']

try:
    df = pd.read_csv('dataset.csv', na_values=valores_nulos_ocultos)
    print(f"📊 Dataset cargado con éxito. Dimensiones iniciales: {df.shape[0]} filas y {df.shape[1]} columnas.")
except FileNotFoundError:
    print("❌ Error: No se encuentra el archivo 'dataset.csv'. Por favor, súbelo a la sesión de Colab.")

# 2.3 Primer vistazo técnico
display(df.head(5))
```

# 2. Auditoría y Calidad de Datos (EDA - Parte 1)

```python
# ==============================================================================
# 2. AUDITORÍA, LIMPIEZA DE COLUMNAS OCULTAS Y CORRECCIÓN DE TIPOS DE DATOS
# ==============================================================================

# 2.1 Inspección de Valores Nulos Reales (después de convertir '?', 'NA', etc.)
nulos_por_columna = df.isnull().sum() #Todas las columnas con sus valores nulos
columnas_con_nulos = nulos_por_columna[nulos_por_columna > 0].sort_values(ascending=False) #Ahora filtramos las columnas

print("🔍 Columnas con valores nulos detectados:")
display(pd.DataFrame({'Nulos': columnas_con_nulos, 'Porcentaje (%)': (columnas_con_nulos / len(df)) * 100}))

# 2.2 CORRECCIÓN CRÍTICA DE TIPOS DE DATOS (Transformando Object a Numérico)
# Creamos una copia limpia para trabajar de manera segura
df_clean = df.copy()

# Función robusta con try-except para limpiar variables de edad y tiempo numérico
def limpiar_columna_numerica(ser):
    """
    Intenta extraer números puros de columnas de texto (ej: '50+' -> 50, 'Under 15' -> 15)
    y convierte la columna a tipo numérico (float).
    """
    try:
        # Extrae la primera secuencia de dígitos que encuentre en el texto
        ser_cleaned = ser.astype(str).str.extract(r'(\d+)')[0]
        return pd.to_numeric(ser_cleaned, errors='coerce')
    except Exception as e:
        print(f"⚠️ Error al limpiar columna: {e}")
        return ser

# Aplicamos la limpieza a las columnas que deberían ser numéricas continuas
cols_a_numerico = ['age', 'years_on_internet']
for col in cols_a_numerico:
    if col in df_clean.columns:
        df_clean[col] = limpiar_columna_numerica(df_clean[col])
        print(f"✅ Columna '{col}' convertida exitosamente a numérico. Tipo actual: {df_clean[col].dtype}")

# 2.3 IMPUTACIÓN ESTRATÉGICA DE NULOS
# Para numéricas: imputamos por la Mediana (resistente a outliers)
for col in cols_a_numerico:
    mediana_val = df_clean[col].median()
    df_clean[col].fillna(mediana_val, inplace=True)

# Para categóricas con nulos críticos (ej: primary_computing_platform con ~26% nulos)
# Creamos una nueva categoría llamada 'Desconocido/No Informado' para no perder 2,700 filas
cols_categoricas_con_nulos = df_clean.select_dtypes(include='object').columns
for col in cols_categoricas_con_nulos:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col].fillna('No_Informado', inplace=True)

print("\n🚀 Limpieza finalizada. Estado actual de nulos en el dataset limpio:")
print(f"Total de nulos restantes en todo el dataframe: {df_clean.isnull().sum().sum()}")
```

# 4. ESTADÍSTICA INFERENCIAL Y VALIDACIÓN DE HIPÓTESIS

```python
# ==============================================================================
# 4. ESTADÍSTICA INFERENCIAL Y VALIDACIÓN DE HIPÓTESIS DE NEGOCIO
# ==============================================================================

print("🎯 INICIANDO VALIDACIÓN ESTADÍSTICA DE HIPÓTESIS DE NEGOCIO...\n")
print("-" * 80)

# ------------------------------------------------------------------------------
# HIPÓTESIS 1: Edad vs. Disposición a Pagar (Kruskal-Wallis / ANOVA)
# ------------------------------------------------------------------------------
print("📌 HIPÓTESIS 1: ¿La edad del usuario influye en su disposición a pagar cuotas web?")
print("   - Variable Numérica: age (Edad)")
print("   - Variable Categórica: willingness_to_pay_fees")

# Separamos los grupos de edad según su respuesta
grupos_pago = [df_clean[df_clean['willingness_to_pay_fees'] == resp]['age'].dropna()
               for resp in df_clean['willingness_to_pay_fees'].unique() if pd.notna(resp)]

# Aplicamos la prueba no paramétrica de Kruskal-Wallis (robusta contra outliers en edad)
stat_kw, p_val_kw = stats.kruskal(*grupos_pago)

print(f"\n📊 Resultados de la prueba Kruskal-Wallis:")
print(f"   - Estadístico de prueba (H): {stat_kw:.4f}")
print(f"   - p-value: {p_val_kw:.4e}")

# Interpretación automática de negocio
alpha = 0.05
if p_val_kw < alpha:
    print("\n✅ CONCLUSIÓN ESTADÍSTICA: RECHAZAMOS la hipótesis nula (p < 0.05).")
    print("💡 INTERPRETACIÓN DE NEGOCIO: Existe una diferencia estadísticamente significativa en la edad")
    print("   de los usuarios según su disposición a pagar. Por lo tanto, SE CONFIRMA la decisión de")
    print("   diseñar planes de precios diferenciados por generación (ej: tarifas preferenciales o freemium para jóvenes).")
else:
    print("\n❌ CONCLUSIÓN ESTADÍSTICA: NO RECHAZAMOS la hipótesis nula (p >= 0.05).")
    print("💡 INTERPRETACIÓN DE NEGOCIO: La edad no determina la disposición a pagar. El modelo de precios")
    print("   debe ser transversal y no enfocarse en criterios etarios.")

print("-" * 80)

# ------------------------------------------------------------------------------
# HIPÓTESIS 2: Nivel Educativo vs. Freno de Compra por Seguridad (Chi-Cuadrado)
# ------------------------------------------------------------------------------
print("📌 HIPÓTESIS 2: ¿El temor a la seguridad web depende del nivel educativo del usuario?")
print("   - Variables Categóricas: education_attainment vs. not_purchasing_security")

# Creamos la tabla de contingencia
tabla_contingencia = pd.crosstab(df_clean['education_attainment'], df_clean['not_purchasing_security'])

# Aplicamos la prueba Chi-Cuadrado de Independencia
stat_chi2, p_val_chi2, dof, expected = stats.chi2_contingency(tabla_contingencia)

print(f"\n📊 Resultados de la prueba Chi-Cuadrado (X²):")
print(f"   - Estadístico de prueba (X²): {stat_chi2:.4f}")
print(f"   - Grados de libertad (dof): {dof}")
print(f"   - p-value: {p_val_chi2:.4e}")

if p_val_chi2 < alpha:
    print("\n✅ CONCLUSIÓN ESTADÍSTICA: RECHAZAMOS la hipótesis nula de independencia (p < 0.05).")
    print("💡 INTERPRETACIÓN DE NEGOCIO: El miedo a la seguridad NO se distribuye por igual; está fuertemente")
    print("   asociado al nivel educativo. SE MODIFICA la estrategia generalista: se debe adaptar el lenguaje")
    print("   técnico de las garantías de seguridad y privacidad según el perfil educativo del segmento.")
else:
    print("\n❌ CONCLUSIÓN ESTADÍSTICA: NO RECHAZAMOS la hipótesis nula (p >= 0.05).")
    print("💡 INTERPRETACIÓN DE NEGOCIO: El temor a la seguridad es universal y no depende del nivel educativo.")

print("-" * 80)
```

# 5. ANÁLISIS MULTIVARIANTE: PCA + CLUSTERING (K-MEANS)

```python
# ==============================================================================
# 5. ANÁLISIS MULTIVARIANTE: REDUCCIÓN DE DIMENSIONALIDAD (PCA) Y CLUSTERING
# ==============================================================================

print("🚀 INICIANDO PIPELINE DE MACHINE LEARNING NO SUPERVISADO...\n")

# 5.1 SELECCIÓN DE VARIABLES CONDUCTUALES Y SOCIODEMOGRÁFICAS PARA EL MODELO
# Seleccionamos edad, años en internet, barreras de compra, membresías y pagadores
cols_clustering = ['age', 'years_on_internet'] + [
    col for col in df_clean.columns if col.startswith('not_purchasing_') or
                                       col.startswith('community_membership_') or
                                       col.startswith('who_pays_')
]

X_raw = df_clean[cols_clustering].copy()

# Estandarización estricta (Vital para que PCA y K-Means funcionen correctamente)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)
print(f"✅ Matriz de características preparada y escalada. Dimensiones: {X_scaled.shape}")

# 5.2 REDUCCIÓN DE DIMENSIONALIDAD CON PCA (Sesión 16)
pca_full = PCA(random_state=42)
pca_full.fit(X_scaled)
varianza_acumulada = np.cumsum(pca_full.explained_variance_ratio_)

# Buscamos cuántos componentes explican al menos el 80% de la varianza
n_componentes_80 = np.argmax(varianza_acumulada >= 0.80) + 1

# Gráfico de Varianza Explicada Acumulada (Scree Plot)
plt.figure(figsize=(10, 5))
plt.plot(range(1, len(varianza_acumulada) + 1), varianza_acumulada, marker='o', color='#2b5c8f', label='Varianza Explicada Acumulada')
plt.axhline(y=0.80, color='r', linestyle='--', label='Umbral del 80% de Varianza')
plt.axvline(x=n_componentes_80, color='g', linestyle='--', label=f'{n_componentes_80} Componentes ({varianza_acumulada[n_componentes_80-1]*100:.1f}%)')
plt.title('PCA - Varianza Explicada Acumulada (Justificación de Dimensionalidad)', fontsize=14, fontweight='bold')
plt.xlabel('Número de Componentes Principales')
plt.ylabel('Proporción de Varianza Explicada')
plt.legend(loc='lower right')
plt.show()

# Aplicamos el PCA definitivo con el número óptimo de componentes
pca_definitivo = PCA(n_components=n_componentes_80, random_state=42)
X_pca = pca_definitivo.fit_transform(X_scaled)
print(f"\n📉 Dimensionalidad reducida de {X_scaled.shape[1]} a {X_pca.shape[1]} componentes principales (Reteniendo 80% de información).\n")

# 5.3 BÚSQUEDA DEL NÚMERO ÓPTIMO DE CLÚSTERS (Método del Codo y Silueta)
rango_k = range(2, 8)
inercia = []
siluetas = []

for k in rango_k:
    kmeans_temp = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans_temp.fit(X_pca)
    inercia.append(kmeans_temp.inertia_)
    siluetas.append(silhouette_score(X_pca, kmeans_temp.labels_))

# Graficamos Codo y Silueta en paralelo
fig, ax1 = plt.subplots(figsize=(12, 5))
color = '#1f77b4'
ax1.set_xlabel('Número de Clústers (k)', fontweight='bold')
ax1.set_ylabel('Inercia (WCSS - Método del Codo)', color=color, fontweight='bold')
ax1.plot(rango_k, inercia, marker='s', color=color, linewidth=2)
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = '#ff7f0e'
ax2.set_ylabel('Coeficiente de Silueta (Más alto es mejor)', color=color, fontweight='bold')
ax2.plot(rango_k, siluetas, marker='o', color=color, linewidth=2, linestyle='--')
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Optimización de Clústers: Método del Codo vs. Coeficiente de Silueta', fontsize=14, fontweight='bold')
plt.show()

# 5.4 MODELADO DEFINITIVO DE PERSONAS DIGITALES (Elegimos k=3 como estándar corporativo)
# Nota: Si el gráfico de silueta te marca que otro k es más alto (ej: 4), puedes cambiar n_clusters=3 por 4.
k_optimo = 4
kmeans_final = KMeans(n_clusters=k_optimo, random_state=42, n_init=20)
df_clean['Cluster_ID'] = kmeans_final.fit_predict(X_pca)

print(f"🏆 Modelo K-Means entrenado con k={k_optimo} clústers.")
print("\n📊 Distribución de usuarios por Clúster:")
print(df_clean['Cluster_ID'].value_counts().sort_index())

# 5.5 PERFILAMIENTO COMERCIAL (Bautizando a las Personas Digitales)
# Calculamos las medias de las variables más representativas para entender quién es cada clúster
resumen_clusters = df_clean.groupby('Cluster_ID')[['age', 'years_on_internet', 'not_purchasing_security', 'not_purchasing_privacy', 'not_purchasing_prefer_people', 'who_pays_for_access_self']].mean()
resumen_clusters['age'] = resumen_clusters['age'].round(1)
resumen_clusters['years_on_internet'] = resumen_clusters['years_on_internet'].round(1)
resumen_clusters.columns = ['Edad Promedio', 'Años Internet', 'Freno: Seguridad', 'Freno: Privacidad', 'Freno: Trato Humano', 'Paga su Conexión (0 a 1)']

print("\n🔍 PERFIL PROMEDIO DE LAS PERSONAS DIGITALES (Radiografía por Clúster):")
display(resumen_clusters)
```

## 6.2 Priorización Estratégica e Insights Accionables

Basado en la evidencia inferencial (pruebas de Kruskal-Wallis y Chi-Cuadrado que validan el impacto de la educación y la edad en el comportamiento de pago y confianza), se priorizan las siguientes acciones:

1. **Estrategia para "Ciudadanos Confiados" (52.3%): Optimización del Embudo de Conversión (UX/UI)**
   * **Evidencia:** No le temen al canal web ni exigen trato humano (apenas un 5% prefiere personas físicas).
   * **Acción:** Implementar flujos de compra "One-Click", venta cruzada automatizada (*Cross-selling*) y checkouts simplificados. No gastar presupuesto publicitario en convencerlos de la seguridad de la página; enfocar el mensaje en precio, catálogo y velocidad de envío.

2. **Estrategia para "Veteranos Escépticos" (26.6%): Blindaje Digital y Marketing de Confianza**
   * **Evidencia:** Representan más de una cuarta parte de los compradores potenciales, tienen los recursos económicos pero presentan un bloqueo casi total por miedo al fraude (84%) y al manejo de sus datos privados (76%).
   * **Acción:** Diseñar una estrategia de *Trust-Building*. Incorporar pasarelas de pago de renombre (PayPal, Apple Pay), visibilizar certificados SSL/ISO en la página de producto, ofrecer garantías de devolución del 100% y redactar políticas de privacidad ultra-transparentes sin letra chica.

3. **Estrategia para "Maduros Pragmáticos" (20.5%): Soporte Híbrido y Simplicidad**
   * **Evidencia:** Segmento de mayor edad al que el modelo tradicional de suscripción rígida no le resulta atractivo.
   * **Acción:** Desplegar canales de soporte asistido en puntos críticos del pago (chatbots eficientes o atención por WhatsApp) para mitigar su fricción en compras complejas, y priorizar modelos de cobro por uso o pasarelas locales familiares en lugar de suscripciones automáticas recurrentes.

4. **Descartar inversión en el Clúster 0 (0.48%): Exclusión de Pauta**
   * **Evidencia:** Al no costear su propia conectividad (sólo 16% paga su acceso), su poder de decisión de compra en e-commerce es nulo o altamente dependiente de terceros.
   * **Acción:** Excluir perfiles sociodemográficos coincidentes en las campañas de marketing pagado (*Paid Media*) para optimizar el Retorno de Inversión (ROI) publicitario.

## 6.3 Limitaciones Metodológicas y Futuras Iteraciones

* **Naturaleza Estática del Dataset:** Los datos corresponden a una fotografía de corte transversal (*cross-sectional*). No capturan la evolución del usuario ni la tasa de adopción tecnológica en el tiempo tras recibir un incentivo comercial.
* **Saturación en Variables Binarias:** Al trabajar con múltiples respuestas *One-Hot Encoding* para los frenos de compra, el modelo es sensible a la co-ocurrencia de respuestas (usuarios que marcan todas las quejas al mismo tiempo).
* **Siguiente Iteración Recomendada:** Ejecutar un experimento de **A/B Testing en entorno real**. Presentar al clúster de "Veteranos Escépticos" una landing page con insignias de ciberseguridad prominentes versus una landing page estándar para medir matemáticamente si la conversión de ese clúster se incrementa con la variable visual de confianza.

## 6.4 Conclusión Final conectada a la Decisión de Negocio

El análisis multivariante ha demostrado que **tratar a los usuarios de internet como una masa homogénea es un error financiero**. La decisión del negocio de lanzar una plataforma universal de E-commerce o suscripción fallará si no resuelve primero el dilema del **Clúster 2 (26.6%)**, donde la pérdida de ventas no es por precio, sino por déficit de confianza digital.
